ТРЕБОВАНИЯ К ФАЙЛУ:
1. Формат: .xlsx (Excel) или .csv
2. Файл должен содержать колонки:
   - 'Спортсмен'  — текстовые метки
   - 'Группа'  — Экспериментальная / Контрольная
   - 'До'  — числовые значения до эксперимента
   - 'После'  — числовые значения после эксперимента

```python
# ============================================================
#  СТАТИСТИЧЕСКИЙ АНАЛИЗ ДЛЯ СПОРТИВНОЙ ПЕДАГОГИКИ
#
# ============================================================

# --- 1. УСТАНОВКА И ИМПОРТ БИБЛИОТЕК ---
!pip install pingouin openpyxl plotly -q

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
from scipy.stats import shapiro, levene, ttest_rel, ttest_ind, mannwhitneyu, wilcoxon
import pingouin as pg
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Настройка стилей
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("viridis")
pd.set_option('display.max_columns', None)

print(" Библиотеки загружены!")
print("="*60)

# ============================================================
# 2. ЗАГРУЗКА ПОЛЬЗОВАТЕЛЬСКОГО ФАЙЛА
# ============================================================
print("\n ЗАГРУЗКА ДАННЫХ")
print("="*60)
print("""
ТРЕБОВАНИЯ К ФАЙЛУ:
1. Формат: .xlsx (Excel) или .csv
2. Файл должен содержать колонки:
   - 'Спортсмен' (или 'Имя', 'ФИО') — текстовые метки
   - 'Группа'  — Экспериментальная / Контрольная
   - 'До' (или 'Before') — числовые значения до эксперимента
   - 'После' (или 'After') — числовые значения после эксперимента
3. Все колонки должны быть заполнены (без пустых ячеек)
""")

from google.colab import files
uploaded = files.upload()

# Определяем имя загруженного файла
file_name = list(uploaded.keys())[0]
print(f"\n Загружен файл: {file_name}")

# Загружаем данные в зависимости от расширения
try:
    if file_name.endswith('.csv'):
        df = pd.read_csv(file_name)
    else:
        df = pd.read_excel(file_name)
    print(" Данные успешно загружены!")
except Exception as e:
    print(f" Ошибка загрузки: {e}")
    print("Попробуйте сохранить файл в формате .xlsx или .csv и загрузить заново.")
    raise

# ============================================================
# 3. ПРЕДВАРИТЕЛЬНАЯ ОБРАБОТКА И ПРОВЕРКА ДАННЫХ
# ============================================================
print("\n" + "="*60)
print("ПРЕДВАРИТЕЛЬНАЯ ПРОВЕРКА ДАННЫХ")
print("="*60)

# Выводим первые строки для проверки
print("\n Первые 5 строк вашего файла:")
display(df.head())

# Проверяем названия колонок и переименовываем при необходимости
print("\n Проверка названий колонок...")

# Словарь для переименования (пользователь может добавить свои варианты)
rename_dict = {}
for col in df.columns:
    col_lower = col.lower().strip()
    if col_lower in ['спортсмен', 'имя', 'фио', 'участник', 'participant', 'name']:
        rename_dict[col] = 'Спортсмен'
    elif col_lower in ['группа', 'group', 'тип', 'type']:
        rename_dict[col] = 'Группа'
    elif col_lower in ['до', 'before', 'pre', 'начало', 'start']:
        rename_dict[col] = 'До'
    elif col_lower in ['после', 'after', 'post', 'конец', 'end']:
        rename_dict[col] = 'После'

if rename_dict:
    df = df.rename(columns=rename_dict)
    print(f" Переименованы колонки: {rename_dict}")
else:
    print(" Переименование не требуется или колонки уже имеют правильные названия")

# Проверяем наличие обязательных колонок
required_cols = ['Спортсмен', 'Группа', 'До', 'После']
missing_cols = [col for col in required_cols if col not in df.columns]

if missing_cols:
    print(f"\n❌ ОШИБКА: Отсутствуют колонки: {missing_cols}")
    print(f"Доступные колонки: {list(df.columns)}")
    print("\nПожалуйста, переименуйте колонки в файле и загрузите заново.")
    print("Требуемые названия: 'Спортсмен', 'Группа', 'До', 'После'")
    raise ValueError("Отсутствуют обязательные колонки")

# Проверка на пустые значения
if df[required_cols].isnull().any().any():
    print("\n ВНИМАНИЕ: Обнаружены пустые значения. Строки с пропусками будут удалены.")
    df = df.dropna(subset=required_cols)
    print(f" После удаления пропусков: {len(df)} строк")

# Проверка уникальных групп
groups = df['Группа'].unique()
print(f"\n Обнаружены группы: {list(groups)}")
if len(groups) != 2:
    print(f" ВНИМАНИЕ: Ожидается 2 группы (Экспериментальная и Контрольная), найдено {len(groups)}")
    print("Убедитесь, что в колонке 'Группа' указаны корректные названия")

# Преобразуем числовые колонки
df['До'] = pd.to_numeric(df['До'], errors='coerce')
df['После'] = pd.to_numeric(df['После'], errors='coerce')

# Удаляем строки с некорректными числами
df = df.dropna(subset=['До', 'После'])

print(f"\n Итоговое количество наблюдений: {len(df)}")

# Базовая статистика
print("\n Общая описательная статистика:")
display(df[['До', 'После']].describe().round(2))

# ============================================================
# 4. ФУНКЦИИ ДЛЯ АНАЛИЗА
# ============================================================

def interpret_effect(d):
    """Интерпретация величины эффекта Коэна"""
    if abs(d) < 0.2:
        return "Очень слабый (практически нет)"
    elif abs(d) < 0.5:
        return "Слабый эффект"
    elif abs(d) < 0.8:
        return "Средний эффект"
    else:
        return "СИЛЬНЫЙ эффект!"

def check_normality(data, group_col, value_col):
    """Проверка нормальности распределения"""
    groups = data[group_col].unique()
    results = []
    for group in groups:
        subset = data[data[group_col] == group][value_col]
        if len(subset) >= 3:
            stat, p = shapiro(subset)
            normal = " ДА" if p > 0.05 else " НЕТ"
            results.append([group, round(stat, 3), round(p, 4), normal])
        else:
            results.append([group, "N/A", "N/A", " Малый размер выборки"])
    return pd.DataFrame(results, columns=['Группа', 'W-статистика', 'p-значение', 'Распределение нормальное?'])

def descriptive_stats(data, group_col, value_cols):
    """Расчет описательной статистики"""
    desc = data.groupby(group_col)[value_cols].agg(['mean', 'std', 'min', 'max', 'count'])
    desc.columns = ['Среднее', 'Ст.Откл.', 'Мин', 'Макс', 'N']
    return desc.round(2)

def compare_paired(before, after, group_name):
    """Сравнение ДО и ПОСЛЕ в одной группе"""
    diff = after - before
    mean_diff = np.mean(diff)
    std_diff = np.std(diff, ddof=1)

    # Проверка нормальности разностей
    if len(diff) >= 3:
        _, p_norm = shapiro(diff)
    else:
        p_norm = 0.01  # Маленькая выборка - используем непараметрику

    if p_norm > 0.05 and len(diff) >= 3:
        # Параметрический тест
        t_stat, p_val = ttest_rel(before, after)
        method = "Парный t-критерий Стьюдента"
        if std_diff != 0:
            cohen_d = abs(mean_diff / std_diff)
        else:
            cohen_d = 0
    else:
        # Непараметрический тест
        try:
            w_stat, p_val = wilcoxon(before, after)
            method = "Критерий Уилкоксона"
            n = len(before)
            mu = n * (n + 1) / 4
            sigma = np.sqrt(n * (n + 1) * (2*n + 1) / 24)
            z_val = (w_stat - mu) / sigma
            cohen_d = abs(z_val / np.sqrt(n))
            if mean_diff < 0:
                cohen_d = -cohen_d
        except:
            # Если тест не удался (например, все разности равны нулю)
            p_val = 1.0
            method = "Невозможно вычислить"
            cohen_d = 0

    return {
        'group': group_name,
        'mean_before': np.mean(before),
        'std_before': np.std(before),
        'mean_after': np.mean(after),
        'std_after': np.std(after),
        'mean_diff': mean_diff,
        'method': method,
        'p_value': p_val,
        'cohen_d': cohen_d,
        'significant': p_val < 0.05,
        'normality_p': p_norm
    }

def compare_independent(group1, group2):
    """Сравнение двух независимых групп"""
    # Проверка нормальности
    if len(group1) >= 3 and len(group2) >= 3:
        _, p_norm1 = shapiro(group1)
        _, p_norm2 = shapiro(group2)
        both_normal = (p_norm1 > 0.05) and (p_norm2 > 0.05)
    else:
        both_normal = False

    if both_normal:
        # Проверка равенства дисперсий
        try:
            _, p_levene = levene(group1, group2)
            if p_levene > 0.05:
                t_stat, p_val = ttest_ind(group1, group2, equal_var=True)
                method = "t-критерий (равные дисперсии)"
            else:
                t_stat, p_val = ttest_ind(group1, group2, equal_var=False)
                method = "t-критерий Уэлча (неравные дисперсии)"
        except:
            p_val = 1.0
            method = "Ошибка вычисления"
    else:
        try:
            u_stat, p_val = mannwhitneyu(group1, group2)
            method = "U-критерий Манна-Уитни"
            # Расчет эффекта через Z
            n1, n2 = len(group1), len(group2)
            mu_u = n1 * n2 / 2
            sigma_u = np.sqrt(n1 * n2 * (n1 + n2 + 1) / 12)
            z_val = (u_stat - mu_u) / sigma_u
            cohen_d = abs(z_val / np.sqrt(n1 + n2))
            return {
                'method': method,
                'mean1': np.mean(group1),
                'std1': np.std(group1),
                'mean2': np.mean(group2),
                'std2': np.std(group2),
                'p_value': p_val,
                'cohen_d': cohen_d,
                'significant': p_val < 0.05
            }
        except:
            p_val = 1.0
            method = "Ошибка вычисления"

    # Расчет Cohen's d для независимых
    n1, n2 = len(group1), len(group2)
    var1, var2 = np.var(group1, ddof=1), np.var(group2, ddof=1)
    pooled_std = np.sqrt(((n1-1)*var1 + (n2-1)*var2) / (n1+n2-2)) if (n1+n2-2) > 0 else 1
    cohen_d = abs(np.mean(group1) - np.mean(group2)) / pooled_std if pooled_std != 0 else 0

    return {
        'method': method,
        'mean1': np.mean(group1),
        'std1': np.std(group1),
        'mean2': np.mean(group2),
        'std2': np.std(group2),
        'p_value': p_val,
        'cohen_d': cohen_d,
        'significant': p_val < 0.05
    }

# ============================================================
# 5. ЭТАП 1: ОПИСАТЕЛЬНАЯ СТАТИСТИКА
# ============================================================
print("\n" + "="*60)
print("ЭТАП 1: ОПИСАТЕЛЬНАЯ СТАТИСТИКА")
print("="*60)

print("\n Описательная статистика ДО эксперимента:")
display(descriptive_stats(df, 'Группа', 'До'))

print("\n Описательная статистика ПОСЛЕ эксперимента:")
display(descriptive_stats(df, 'Группа', 'После'))

# Boxplot
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
sns.boxplot(x='Группа', y='До', data=df, palette='Set2', ax=axes[0])
axes[0].set_title('Распределение ДО эксперимента', fontsize=14)
axes[0].set_ylabel('Показатель')
sns.boxplot(x='Группа', y='После', data=df, palette='Set3', ax=axes[1])
axes[1].set_title('Распределение ПОСЛЕ эксперимента', fontsize=14)
axes[1].set_ylabel('Показатель')
plt.tight_layout()
plt.show()

# ============================================================
# 6. ЭТАП 2: ПРОВЕРКА НОРМАЛЬНОСТИ
# ============================================================
print("\n" + "="*60)
print("ЭТАП 2: ПРОВЕРКА НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ")
print("="*60)

print("\n Для показателя 'До':")
display(check_normality(df, 'Группа', 'До'))

print("\n Для показателя 'После':")
display(check_normality(df, 'Группа', 'После'))

# Q-Q Plot
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
groups = df['Группа'].unique()
if len(groups) >= 2:
    titles = [
        f'До - {groups[0]}',
        f'До - {groups[1]}',
        f'После - {groups[0]}',
        f'После - {groups[1]}'
    ]
    axes = axes.flatten()
    plot_idx = 0
    for group in groups:
        for time_point in ['До', 'После']:
            subset = df[df['Группа'] == group][time_point]
            if len(subset) >= 3:
                stats.probplot(subset, dist="norm", plot=axes[plot_idx])
                axes[plot_idx].set_title(titles[plot_idx])
            else:
                axes[plot_idx].text(0.5, 0.5, f'Недостаточно данных\nдля {titles[plot_idx]}',
                                   ha='center', va='center', transform=axes[plot_idx].transAxes)
            plot_idx += 1
plt.tight_layout()
plt.show()

# ============================================================
# 7. ЭТАП 3: СРАВНЕНИЕ ДО И ПОСЛЕ
# ============================================================
print("\n" + "="*60)
print("ЭТАП 3: СРАВНЕНИЕ 'ДО' и 'ПОСЛЕ' В КАЖДОЙ ГРУППЕ")
print("="*60)

# Отделяем группы
group_names = df['Группа'].unique()
results_paired = {}

for group in group_names:
    before = df[df['Группа'] == group]['До']
    after = df[df['Группа'] == group]['После']
    if len(before) >= 3:
        results_paired[group] = compare_paired(before, after, group)
    else:
        print(f" Группа '{group}' имеет слишком мало наблюдений ({len(before)}) для анализа")

# Вывод результатов
for group, res in results_paired.items():
    print(f"\n {res['group']} группа:")
    print(f"   Метод: {res['method']}")
    print(f"   ДО: {res['mean_before']:.2f} ± {res['std_before']:.2f}")
    print(f"   ПОСЛЕ: {res['mean_after']:.2f} ± {res['std_after']:.2f}")
    print(f"   Прирост: {res['mean_diff']:.2f}")
    print(f"   p-значение: {res['p_value']:.4f} {' Значимо!' if res['significant'] else '❌ Не значимо'}")
    print(f"    Cohen's d = {res['cohen_d']:.3f} ({interpret_effect(res['cohen_d'])})")

# ============================================================
# 8. ЭТАП 4: СРАВНЕНИЕ ГРУПП МЕЖДУ СОБОЙ
# ============================================================
print("\n" + "="*60)
print("ЭТАП 4: СРАВНЕНИЕ ГРУПП МЕЖДУ СОБОЙ")
print("="*60)

if len(group_names) >= 2:
    group1 = df[df['Группа'] == group_names[0]]['После']
    group2 = df[df['Группа'] == group_names[1]]['После']

    if len(group1) >= 3 and len(group2) >= 3:
        res_ind = compare_independent(group1, group2)

        print(f"\n Сравнение групп ПОСЛЕ эксперимента:")
        print(f"   Метод: {res_ind['method']}")
        print(f"   {group_names[0]}: {res_ind['mean1']:.2f} ± {res_ind['std1']:.2f}")
        print(f"   {group_names[1]}: {res_ind['mean2']:.2f} ± {res_ind['std2']:.2f}")
        print(f"   Разница: {res_ind['mean1'] - res_ind['mean2']:.2f}")
        print(f"   p-значение: {res_ind['p_value']:.4f} {' Значимо!' if res_ind['significant'] else '❌ Не значимо'}")
        print(f"    Cohen's d = {res_ind['cohen_d']:.3f} ({interpret_effect(res_ind['cohen_d'])})")
    else:
        print(f" Недостаточно данных для сравнения групп")

# ============================================================
# 9. ЭТАП 5: КОРРЕЛЯЦИОННЫЙ АНАЛИЗ
# ============================================================
print("\n" + "="*60)
print("ЭТАП 5: КОРРЕЛЯЦИОННЫЙ АНАЛИЗ")
print("="*60)

if len(df) >= 3:
    _, p_norm_d = shapiro(df['До'])
    _, p_norm_a = shapiro(df['После'])

    if p_norm_d > 0.05 and p_norm_a > 0.05:
        corr_method = "Пирсон"
        corr_matrix = df[['До', 'После']].corr(method='pearson')
    else:
        corr_method = "Спирмен"
        corr_matrix = df[['До', 'После']].corr(method='spearman')

    print(f" Метод корреляции: {corr_method}")
    display(corr_matrix.round(3))

    # Scatter plot с линией регрессии
    fig, ax = plt.subplots(figsize=(10, 6))
    sns.regplot(x='До', y='После', data=df, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
    ax.set_title(f'Корреляция показателей ({corr_method})', fontsize=14)
    r = corr_matrix.iloc[0, 1]
    ax.text(0.05, 0.95, f'R² = {r**2:.3f}', transform=ax.transAxes, fontsize=12,
            verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    plt.grid(True, alpha=0.3)
    plt.show()

    print(f"\n Коэффициент корреляции: r = {r:.3f}")
    print(f" Коэффициент детерминации R² = {r**2:.3f}")
else:
    print(" Недостаточно данных для корреляционного анализа")

# ============================================================
# 10. ЭТАП 6: ИНТЕРАКТИВНЫЙ ГРАФИК
# ============================================================
print("\n" + "="*60)
print("ЭТАП 6: ИНДИВИДУАЛЬНАЯ ДИНАМИКА")
print("="*60)

fig_ind = go.Figure()

# Определяем цвета для групп
group_colors = {}
palette = ['#FF4444', '#4444FF', '#44AA44', '#FFAA00']
for i, group in enumerate(group_names):
    group_colors[group] = palette[i % len(palette)]

for i, row in df.iterrows():
    color = group_colors.get(row['Группа'], '#888888')
    fig_ind.add_trace(go.Scatter(
        x=['До', 'После'],
        y=[row['До'], row['После']],
        mode='lines+markers',
        name=row['Спортсмен'],
        line=dict(color=color, width=1.5),
        marker=dict(size=8),
        text=f"{row['Спортсмен']}<br>Группа: {row['Группа']}<br>Прирост: {row['После']-row['До']:.2f}",
        hoverinfo='text',
        showlegend=False
    ))

# Добавляем легенду для групп
for group in group_names:
    fig_ind.add_trace(go.Scatter(
        x=[None], y=[None],
        mode='markers',
        marker=dict(size=10, color=group_colors.get(group, '#888888')),
        name=group,
        showlegend=True
    ))

fig_ind.update_layout(
    title='Индивидуальная динамика каждого спортсмена',
    xaxis_title='Время тестирования',
    yaxis_title='Значение показателя',
    height=600,
    hovermode='closest',
    legend_title='Группа'
)
fig_ind.show()

# ============================================================
# 11. ДОПОЛНИТЕЛЬНАЯ ВИЗУАЛИЗАЦИЯ: СРАВНЕНИЕ ПРИРОСТОВ
# ============================================================
print("\n" + "="*60)
print("ДОПОЛНИТЕЛЬНАЯ ВИЗУАЛИЗАЦИЯ: ПРИРОСТЫ ГРУПП")
print("="*60)

df['Прирост'] = df['После'] - df['До']

fig_compare = go.Figure()
for group in group_names:
    subset = df[df['Группа'] == group]
    fig_compare.add_trace(go.Box(
        y=subset['Прирост'],
        name=group,
        boxmean='sd',
        marker_color=group_colors.get(group, '#888888'),
        boxpoints='all',
        jitter=0.3,
        pointpos=-1.8
    ))

fig_compare.update_layout(
    title='Сравнение приростов между группами',
    yaxis_title='Прирост показателя',
    height=500,
    hovermode='closest'
)
fig_compare.show()

# ============================================================
# 12. ИТОГОВОЕ ЗАКЛЮЧЕНИЕ
# ============================================================
print("\n" + "="*60)
print("ИТОГОВОЕ ЗАКЛЮЧЕНИЕ ПО РЕЗУЛЬТАТАМ ИССЛЕДОВАНИЯ")
print("="*60)

print(f"""
 АНАЛИТИЧЕСКИЙ ОТЧЕТ:

1. Объем выборки: N = {len(df)} спортсменов
2. Количество групп: {len(group_names)} ({', '.join(group_names)})
""")

for group, res in results_paired.items():
    print(f"""
3. Группа '{group}':
   - Прирост: {res['mean_diff']:.2f}
   - p = {res['p_value']:.4f} {'' if res['significant'] else ''}
   - Cohen's d = {res['cohen_d']:.3f} ({interpret_effect(res['cohen_d'])})
""")

if len(group_names) >= 2 and 'res_ind' in locals():
    print(f"""
4. Межгрупповое сравнение:
   - Разница между группами: {res_ind['mean1'] - res_ind['mean2']:.2f}
   - p = {res_ind['p_value']:.4f} {'✅' if res_ind['significant'] else '❌'}
   - Cohen's d = {res_ind['cohen_d']:.3f} ({interpret_effect(res_ind['cohen_d'])})
""")

if 'r' in locals():
    print(f"""
5. Корреляционная связь:
   - r = {r:.3f} (R² = {r**2:.3f})
""")

print("="*60)
print("\n АНАЛИЗ  ЗАВЕРШЕН!")
```

```python
# =============================================================================
# УНИВЕРСАЛЬНЫЙ СТАТИСТИЧЕСКИЙ АНАЛИЗ: ДО vs ПОСЛЕ
# Для файлов Excel со столбцами: Спортсмен | Группа | До | После
# Адаптировано для Google Colab
# =============================================================================
# Описание: Полный статистический анализ с подробными пояснениями,
#           автоматическим выбором критериев, расчетом эффектов и визуализацией.
#           Каждый график сохраняется отдельным файлом.
#
# =============================================================================

# ═════════════════════════════════════════════════════════════════════════════
# 0. УСТАНОВКА БИБЛИОТЕК (раскомментируйте в Google Colab при необходимости)
# ═════════════════════════════════════════════════════════════════════════════
# !pip install pandas numpy matplotlib seaborn scipy openpyxl statsmodels -q

# ═════════════════════════════════════════════════════════════════════════════
# 1. ИМПОРТ БИБЛИОТЕК
# ═════════════════════════════════════════════════════════════════════════════
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import (
    shapiro, normaltest, levene,
    ttest_rel, ttest_ind,
    wilcoxon, mannwhitneyu,
    f_oneway, pearsonr
)
import warnings
import os
import sys
warnings.filterwarnings('ignore')

# Попытка импорта statsmodels для ANOVA/ANCOVA
try:
    import statsmodels.api as sm
    import statsmodels.formula.api as smf
    from statsmodels.stats.diagnostic import het_breuschpagan
    STATSMODELS_AVAILABLE = True
except ImportError:
    STATSMODELS_AVAILABLE = False

# Для красивых таблиц в Colab
try:
    from IPython.display import display, HTML
    IPYTHON_AVAILABLE = True
except ImportError:
    IPYTHON_AVAILABLE = False

# ═════════════════════════════════════════════════════════════════════════════
# 2. НАСТРОЙКА ВИЗУАЛИЗАЦИИ
# ════════════════════════════════════════════════════════════════════════════
plt.rcParams['figure.dpi'] = 120
plt.rcParams['savefig.dpi'] = 150
plt.rcParams['font.size'] = 11
plt.rcParams['axes.titlesize'] = 13
plt.rcParams['axes.labelsize'] = 11
sns.set_style("whitegrid")
sns.set_context("notebook", font_scale=1.1)

PALETTE = {
    "Контрольная": "#3498db",
    "Экспериментальная": "#e74c3c",
    "До": "#9b59b6",
    "После": "#2ecc71",
    "Изменение": "#f39c12"
}

# ═════════════════════════════════════════════════════════════════════════════
# 3. ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ
# ═════════════════════════════════════════════════════════════════════════════

def print_section(title, symbol="="):
    """ заголовок раздела."""
    line = symbol * 78
    print("\n" + line)
    print(f"  {title}")
    print(line + "\n")


def print_subsection(title):
    """Подзаголовок."""
    print(f"\n{'─' * 78}")
    print(f" {title}")
    print("─" * 78 + "\n")


def print_explanation(text):
    """Блок с пояснением для неподготовленного пользователя."""
    for line in text.split("\n"):
        print(f"     {line}")
    print()


def color_pvalue(val):
    """Цветовое форматирование p-value для applymap."""
    try:
        p = float(val)
        if p < 0.001:
            return 'background-color: #27ae60; color: white; font-weight: bold'
        elif p < 0.01:
            return 'background-color: #2ecc71; color: white; font-weight: bold'
        elif p < 0.05:
            return 'background-color: #f1c40f; color: black; font-weight: bold'
        else:
            return 'background-color: #e74c3c; color: white'
    except:
        return ''


def show_table(df, title=None, gradient_cols=None, pvalue_cols=None,
             cmap='YlGnBu', precision=3):
    """
    Выводит красивую стилизованную таблицу DataFrame через pandas Styler.

    Параметры:
      df            — DataFrame для отображения
      title         — заголовок таблицы
      gradient_cols — список колонок для градиентной подсветки
      pvalue_cols   — список колонок p-value для цветового кодирования
      cmap          — цветовая карта для градиента
      precision     — число знаков после запятой
    """
    if title:
        print(f"\n {title}\n")

    if not IPYTHON_AVAILABLE:
        # Консольный вывод
        print(df.to_string(index=False))
        print()
        return

    # Создаем копию
    df_disp = df.copy()

    # Форматируем числовые колонки
    for col in df_disp.columns:
        if df_disp[col].dtype in ['float64', 'int64']:
            df_disp[col] = df_disp[col].round(precision)

    styler = df_disp.style

    # --- 1. Общие свойства ячеек ---
    styler = styler.set_properties(**{
        'text-align': 'center',
        'font-size': '13px',
        'border': '1px solid #bdc3c7'
    })

    # --- 2. Стили заголовков и строк ---
    styler = styler.set_table_styles([
        {'selector': 'th',
         'props': [
             ('background-color', '#2c3e50'),
             ('color', 'white'),
             ('font-weight', 'bold'),
             ('text-align', 'center'),
             ('padding', '10px'),
             ('border', '1px solid #7f8c8d'),
             ('font-size', '13px')
         ]},
        {'selector': 'td',
         'props': [
             ('padding', '8px'),
             ('border', '1px solid #bdc3c7')
         ]},
        {'selector': 'tr:nth-child(even)',
         'props': [('background-color', '#f8f9fa')]},
        {'selector': 'tr:hover',
         'props': [('background-color', '#d5dbdb')]}
    ])

    # --- 3. Градиентная подсветка числовых колонок ---
    if gradient_cols:
        numeric_grad = [c for c in gradient_cols
                        if c in df_disp.columns
                        and df_disp[c].dtype in ['float64', 'int64']]
        if numeric_grad:
            styler = styler.background_gradient(
                subset=numeric_grad, cmap=cmap, axis=0
            )

    # --- 4. Цветовое кодирование p-value ---
    if pvalue_cols:
        for col in pvalue_cols:
            if col in df_disp.columns:
                styler = styler.applymap(color_pvalue, subset=[col])

    # --- 5. Форматирование и скрытие индекса ---
    styler = styler.format(precision=precision)
    styler = styler.hide(axis='index')

    display(HTML(styler.to_html()))
    print()


def interpret_pvalue(p):
    """Интерпретация p-value со звездочками."""
    if p < 0.001:
        return f"{p:.4f} *** (высоко значимо)"
    elif p < 0.01:
        return f"{p:.4f} ** (значимо)"
    elif p < 0.05:
        return f"{p:.4f} * (достоверно)"
    else:
        return f"{p:.4f} (НЕДОСТОВЕРНО)"


def interpret_cohens_d(d):
    """Интерпретация Cohen's d по Cohen (1988)."""
    a = abs(d)
    if a < 0.2:
        return "незначительный"
    elif a < 0.5:
        return "малый"
    elif a < 0.8:
        return "средний"
    else:
        return "большой"


def interpret_cliffs_delta(d):
    """Интерпретация Cliff's delta."""
    a = abs(d)
    if a < 0.147:
        return "незначительный"
    elif a < 0.33:
        return "малый"
    elif a < 0.474:
        return "средний"
    else:
        return "большой"


def interpret_r_rosenthal(r):
    """Интерпретация r Розенталя."""
    a = abs(r)
    if a < 0.1:
        return "незначительный"
    elif a < 0.3:
        return "малый"
    elif a < 0.5:
        return "средний"
    else:
        return "большой"


def cohens_d_paired(before, after):
    """Cohen's d для парных выборок."""
    diff = np.array(after) - np.array(before)
    m_diff = np.mean(diff)
    sd_diff = np.std(diff, ddof=1)
    return m_diff / sd_diff if sd_diff != 0 else 0.0


def cohens_d_independent(x1, x2):
    """Cohen's d для независимых выборок."""
    x1, x2 = np.array(x1), np.array(x2)
    n1, n2 = len(x1), len(x2)
    if n1 < 2 or n2 < 2:
        return np.nan
    v1 = np.var(x1, ddof=1)
    v2 = np.var(x2, ddof=1)
    pooled = np.sqrt(((n1 - 1) * v1 + (n2 - 1) * v2) / (n1 + n2 - 2))
    return (np.mean(x1) - np.mean(x2)) / pooled if pooled != 0 else 0.0


def cliffs_delta(x, y):
    """Cliff's delta для независимых выборок."""
    x, y = np.array(x), np.array(y)
    nx, ny = len(x), len(y)
    dom = 0
    for xi in x:
        for yi in y:
            if xi > yi:
                dom += 1
            elif xi < yi:
                dom -= 1
    return dom / (nx * ny)


def rosenthals_r_from_wilcoxon(w_stat, n):
    """r Розенталя для зависимых выборок."""
    mean_w = n * (n + 1) / 4
    sd_w = np.sqrt(n * (n + 1) * (2 * n + 1) / 24)
    if sd_w == 0:
        return 0.0, 0.0
    z = (w_stat - mean_w) / sd_w
    r = z / np.sqrt(n)
    return r, z


def bootstrap_median_ci(data, n_boot=2000, ci=95):
    """95% доверительный интервал для медианы методом бутстрапа."""
    np.random.seed(42)
    boot_medians = [np.median(np.random.choice(data, size=len(data), replace=True))
                    for _ in range(n_boot)]
    lower = np.percentile(boot_medians, (100 - ci) / 2)
    upper = np.percentile(boot_medians, 100 - (100 - ci) / 2)
    return lower, upper


def save_figure(name):
    """Сохраняет текущий график в файл и показывает его."""
    fname = f"{name}.png"
    plt.savefig(fname, dpi=150, bbox_inches='tight', facecolor='white')
    print(f"     График сохранен: {fname}")
    plt.show()
    plt.close()


# ════════════════════════════════════════════════════════════════════════════
# 4. ЗАГРУЗКА ФАЙЛА (с диска или по пути)
# ═════════════════════════════════════════════════════════════════════════════
print_section("ЗАГРУЗКА ДАННЫХ")



file_path = None


try:
    from google.colab import files
    print("     Пожалуйста, выберите файл Excel на вашем компьютере...")
    uploaded = files.upload()
    file_path = list(uploaded.keys())[0]
    print(f"    ✅ Файл загружен: {file_path}")
except Exception as e:
      print(f"    ❌ Ошибка загрузки: {e}")
      print("    Попробуйте вариант 2 - укажите путь вручную.")
      file_path = input("    Введите путь к файлу: ").strip()


if not os.path.exists(file_path):
    print(f"\n❌ Файл не найден: {file_path}")
    print("    Пожалуйста, убедитесь, что файл загружен в текущую директорию.")
    sys.exit()

# Чтение данных
df = pd.read_excel(file_path)

# Проверка обязательных столбцов
REQUIRED = ['Спортсмен', 'Группа', 'До', 'После']
missing = [c for c in REQUIRED if c not in df.columns]
if missing:
    print(f"\n❌ В файле отсутствуют столбцы: {missing}")
    print(f"    Найдены столбцы: {list(df.columns)}")
    sys.exit()

# Расчет изменений
df['Изменение'] = df['После'] - df['До']
df['Изменение_%'] = (df['После'] - df['До']) / df['До'] * 100

# Автоопределение групп
group_names = sorted(df['Группа'].unique())
if len(group_names) != 2:
    print(f"\n❌ Ожидалось 2 группы, найдено {len(group_names)}: {group_names}")
    sys.exit()

GROUP_CTRL = [g for g in group_names if 'контроль' in g.lower() or 'control' in g.lower()]
GROUP_EXP = [g for g in group_names if 'эксперимент' in g.lower() or 'experiment' in g.lower()]

if not GROUP_CTRL or not GROUP_EXP:
    GROUP_CTRL = [group_names[0]]
    GROUP_EXP = [group_names[1]]
    print("    ⚠ Не удалось распознать группы по названию. Первая группа принята как Контрольная, вторая как Экспериментальная.")

GROUP_CTRL = GROUP_CTRL[0]
GROUP_EXP = GROUP_EXP[0]

ctrl_df = df[df['Группа'] == GROUP_CTRL].copy().reset_index(drop=True)
exp_df = df[df['Группа'] == GROUP_EXP].copy().reset_index(drop=True)

n_ctrl = len(ctrl_df)
n_exp = len(exp_df)

print(f"""
╔════════════════════════════════════════════════════════════════════════════╗
║  ДАННЫЕ УСПЕШНО ЗАГРУЖЕНЫ                                              ║
╠═════════════════════════════════════════════════════════════════════════════╣
║  Файл: {file_path:<67}║
║  Всего спортсменов: {len(df):<52}║
║  Контрольная группа  ("{GROUP_CTRL}"): n = {n_ctrl:<43}║
║  Экспериментальная   ("{GROUP_EXP}"): n = {n_exp:<43}║
╚═════════════════════════════════════════════════════════════════════════════╝
""")

# ═════════════════════════════════════════════════════════════════════════════
# 5. ОПИСАТЕЛЬНАЯ СТАТИСТИКА
# ═════════════════════════════════════════════════════════════════════════════
print_section("1. ОПИСАТЕЛЬНАЯ СТАТИСТИКА")
df.describe()
ctrl_df.describe()
exp_df.describe()
print_explanation(
    "Здесь мы смотрим на 'портрет' данных до начала сложных расчетов.\n"
    "M = среднее арифметическое, SD = стандартное отклонеие,\n"
    "Md = медиана, IQR = межквартильный размах [Q1; Q3], SE = стандартная ошибка."
)

desc_rows = []
for g_name, g_df, label in [(GROUP_CTRL, ctrl_df, "Контрольная"),
                              (GROUP_EXP, exp_df, "Экспериментальная")]:
    print_subsection(f"{label} группа (n = {len(g_df)})")
    for col in ['До', 'После', 'Изменение']:
        data = g_df[col].dropna()
        m = data.mean()
        sd = data.std(ddof=1)
        se = sd / np.sqrt(len(data))
        md = data.median()
        q25 = data.quantile(0.25)
        q75 = data.quantile(0.75)

        print(f"  {col:12} | M = {m:7.2f} | SD = {sd:6.2f} | SE = {se:5.2f} | "
              f"Md = {md:7.2f} | IQR = [{q25:7.2f}, {q75:7.2f}]")

        desc_rows.append({
            'Группа': label,
            'Показатель': col,
            'M (среднее)': round(m, 2),
            'SD': round(sd, 2),
            'SE': round(se, 2),
            'Md (медиана)': round(md, 2),
            'IQR нижн. (Q1)': round(q25, 2),
            'IQR верхн. (Q3)': round(q75, 2)
        })
    print()

desc_table = pd.DataFrame(desc_rows)
show_table(desc_table,
           title="СВОДНАЯ ТАБЛИЦА ОПИСАТЕЛЬНОЙ СТАТИСТИКИ",
           gradient_cols=['M (среднее)', 'Md (медиана)'],
           cmap='Blues',
           precision=2)

# ═════════════════════════════════════════════════════════════════════════════
# 6. ПРОВЕРКА НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ
# ═════════════════════════════════════════════════════════════════════════════
print_section("2. ПРОВЕРКА НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ")

print_explanation(
    "H0: данные распределены нормально. Если p < 0.05 - отвергаем H0,\n"
    "значит распределение НЕ нормальное."
)

normality_rows = []
all_normal = True

for g_label, g_df in [("Контрольная", ctrl_df), ("Экспериментальная", exp_df)]:
    for col in ['До', 'После', 'Изменение']:
        data = g_df[col].dropna().values
        n = len(data)

        stat_sw, p_sw = shapiro(data)
        stat_dp, p_dp = normaltest(data)
        skew = stats.skew(data)
        kurt = stats.kurtosis(data, fisher=True)

        is_normal = (p_sw > 0.05) and (p_dp > 0.05)
        if not is_normal:
            all_normal = False

        status = "✅ НОРМАЛЬНОЕ" if is_normal else "❌ НЕ НОРМАЛЬНОЕ"

        print(f"{g_label:16} | {col:10} | n={n:2} | "
              f"Shapiro W={stat_sw:.4f}, p={p_sw:.4f} | {status}")
        print(f"                   Скошенность = {skew:7.3f} | Эксцесс = {kurt:7.3f}")
        print()

        normality_rows.append({
            'Группа': g_label,
            'Показатель': col,
            'n': n,
            'Shapiro W': round(stat_sw, 4),
            'Shapiro p': round(p_sw, 4),
            'DP K2': round(stat_dp, 3),
            'DP p': round(p_dp, 4),
            'Скошенность': round(skew, 3),
            'Эксцесс': round(kurt, 3),
            'Вывод': 'Норма' if is_normal else 'Не норма'
        })

norm_table = pd.DataFrame(normality_rows)
show_table(norm_table,
           title="СВОДНАЯ ТАБЛИЦА ПРОВЕРКИ НОРМАЛЬНОСТИ",
           gradient_cols=['Shapiro W', 'DP K2'],
           pvalue_cols=['Shapiro p', 'DP p'],
           cmap='RdYlGn_r',
           precision=4)

print("▶ ОБЩИЙ ВЫВОД:")
if all_normal:
    print("    ✅ Все распределения можно считать нормальными.")
    print("    → Будем использовать ПАРАМЕТРИЧЕСКИЕ критерии (t-тесты).")
else:
    print("    ❌ Не все распределения нормальны.")
    print("    → Будем использовать НЕПАРАМЕТРИЧЕСКИЕ критерии (Wilcoxon, Mann-Whitney).")
print()

# ═════════════════════════════════════════════════════════════════════════════
# 7. ВНУТРИГРУППОВЫЕ СРАВНЕНИЯ (До vs После)
# ═════════════════════════════════════════════════════════════════════════════
print_section("3. ВНУТРИГРУППОВЫЕ СРАВНЕНИЯ: ДО vs ПОСЛЕ")

print_explanation(
    "H0: различий между До и После нет. Если p < 0.05 - отклоняем H0."
)

within_rows = []

for g_label, g_df in [("Контрольная", ctrl_df), ("Экспериментальная", exp_df)]:
    print_subsection(f"{g_label} группа")

    before = g_df['До'].dropna().values
    after = g_df['После'].dropna().values
    diff = after - before
    n_g = len(before)

    md_b = np.median(before)
    md_a = np.median(after)
    iqr_b = (np.percentile(before, 25), np.percentile(before, 75))
    iqr_a = (np.percentile(after, 25), np.percentile(after, 75))

    _, p_sw_diff = shapiro(diff)

    print(f"  Медиана До:    Md = {md_b:.2f}, IQR = [{iqr_b[0]:.2f}, {iqr_b[1]:.2f}]")
    print(f"  Медиана После: Md = {md_a:.2f}, IQR = [{iqr_a[0]:.2f}, {iqr_a[1]:.2f}]")
    print(f"  Сдвиг медианы: ΔMd = {md_a - md_b:+.2f}")
    print()

    if p_sw_diff > 0.05:
        stat_t, p_t = ttest_rel(before, after)
        d = cohens_d_paired(before, after)
        r_ros = stat_t / np.sqrt(stat_t**2 + (n_g - 1)) if stat_t != 0 else 0.0
        direction = "рост" if md_a > md_b else "снижение"

        print("  ✅ Распределение разностей НОРМАЛЬНО.")
        print("  → Парный t-тест (параметрический).")
        print(f"  Результат: t = {stat_t:.3f}, {interpret_pvalue(p_t)}")
        print(f"  Cohen's d = {d:.3f} - {interpret_cohens_d(d)} эффект")
        print(f"  Rosenthal's r = {r_ros:.3f} - {interpret_r_rosenthal(r_ros)} эффект")
        print(f"  Направление: {direction} показателей")
        print()

        within_rows.append({
            'Группа': g_label,
            'Критерий': 'Парный t-тест',
            'Статистика': f"t = {stat_t:.3f}",
            'p-value': round(p_t, 4),
            'Размер эффекта': f"Cohen's d = {d:.3f} ({interpret_cohens_d(d)})",
            'Направление': direction
        })
    else:
        stat_w, p_w = wilcoxon(before, after)
        r_ros, z_val = rosenthals_r_from_wilcoxon(stat_w, n_g)
        direction = "рост" if md_a > md_b else "снижение"

        print("  ❌ Распределение разностей НЕ НОРМАЛЬНО.")
        print("  → Wilcoxon signed-rank (непараметрический).")
        print(f"  Результат: W = {stat_w:.3f}, {interpret_pvalue(p_w)}")
        print(f"  Z = {z_val:.3f}")
        print(f"  Rosenthal's r = {r_ros:.3f} - {interpret_r_rosenthal(r_ros)} эффект")
        print(f"  Направление: {direction} показателей")
        print()

        within_rows.append({
            'Группа': g_label,
            'Критерий': 'Wilcoxon signed-rank',
            'Статистика': f"W = {stat_w:.3f}",
            'p-value': round(p_w, 4),
            'Размер эффекта': f"Rosenthal's r = {r_ros:.3f} ({interpret_r_rosenthal(r_ros)})",
            'Направление': direction
        })

within_table = pd.DataFrame(within_rows)
show_table(within_table,
           title="СВОДНАЯ ТАБЛИЦА ВНУТРИГРУППОВЫХ СРАВНЕНИЙ",
           pvalue_cols=['p-value'],
           cmap='RdYlGn_r',
           precision=4)

# ═════════════════════════════════════════════════════════════════════════════
# 8. МЕЖГРУППОВЫЕ СРАВНЕНИЯ
# ═════════════════════════════════════════════════════════════════════════════
print_section("4. МЕЖГРУППОВЫЕ СРАВНЕНИЯ")

print_explanation(
    "Сравниваем группы МЕЖДУ СОБОЙ. H0: группы не различаются."
)

between_rows = []

# 8a. Сравнение ИЗМЕНЕНИЙ
print_subsection("4a. Сравнение величины изменений: Экспериментальная vs Контрольная")

exp_change = exp_df['Изменение'].dropna().values
ctrl_change = ctrl_df['Изменение'].dropna().values

_, p_sw_ec = shapiro(exp_change)
_, p_sw_cc = shapiro(ctrl_change)
stat_lev, p_lev = levene(exp_change, ctrl_change)

md_ec = np.median(exp_change)
md_cc = np.median(ctrl_change)
iqr_ec = (np.percentile(exp_change, 25), np.percentile(exp_change, 75))
iqr_cc = (np.percentile(ctrl_change, 25), np.percentile(ctrl_change, 75))

print(f"  Экспериментальная: Md изменения = {md_ec:.2f}, IQR = [{iqr_ec[0]:.2f}, {iqr_ec[1]:.2f}]")
print(f"  Контрольная:       Md изменения = {md_cc:.2f}, IQR = [{iqr_cc[0]:.2f}, {iqr_cc[1]:.2f}]")
print(f"  Разница медиан:    ΔMd = {md_ec - md_cc:+.2f}")
print()
print(f"  Levene test: W = {stat_lev:.4f}, {interpret_pvalue(p_lev)}")
print()

if (p_sw_ec > 0.05) and (p_sw_cc > 0.05):
    equal_var = (p_lev > 0.05)
    stat_t, p_t = ttest_ind(exp_change, ctrl_change, equal_var=equal_var)
    d = cohens_d_independent(exp_change, ctrl_change)
    test_name = "Независимый t-тест" + (" (Классический)" if equal_var else " (Уэлча)")

    print("  ✅ Распределения изменений нормальны.")
    print(f"  → {test_name}.")
    print(f"  Результат: t = {stat_t:.3f}, {interpret_pvalue(p_t)}")
    print(f"  Cohen's d = {d:.3f} - {interpret_cohens_d(d)} эффект")
    print()

    between_rows.append({
        'Сравнение': 'Изменения (После-До)',
        'Критерий': test_name,
        'Статистика': f"t = {stat_t:.3f}",
        'p-value': round(p_t, 4),
        'Размер эффекта': f"Cohen's d = {d:.3f} ({interpret_cohens_d(d)})",
        'Вывод': 'Различия значимы' if p_t < 0.05 else 'Различия НЕ значимы'
    })
else:
    stat_u, p_u = mannwhitneyu(exp_change, ctrl_change, alternative='two-sided')
    d_cliff = cliffs_delta(exp_change, ctrl_change)
    direction = "большее улучшение в экспериментальной" if md_ec > md_cc else "большее улучшение в контрольной"

    print("  ❌ Распределения изменений НЕ нормальны.")
    print("  → Mann-Whitney U (непараметрический).")
    print(f"  Результат: U = {stat_u:.3f}, {interpret_pvalue(p_u)}")
    print(f"  Cliff's delta = {d_cliff:.3f} - {interpret_cliffs_delta(d_cliff)} эффект")
    print(f"  Направление: {direction}")
    print()

    between_rows.append({
        'Сравнение': 'Изменения (После-До)',
        'Критерий': 'Mann-Whitney U',
        'Статистика': f"U = {stat_u:.3f}",
        'p-value': round(p_u, 4),
        'Размер эффекта': f"Cliff's delta = {d_cliff:.3f} ({interpret_cliffs_delta(d_cliff)})",
        'Вывод': 'Различия значимы' if p_u < 0.05 else 'Различия НЕ значимы'
    })

# 8b. Сравнение результатов ПОСЛЕ
print_subsection("4b. Сравнение результатов 'После': Экспериментальная vs Контрольная")

exp_after = exp_df['После'].dropna().values
ctrl_after = ctrl_df['После'].dropna().values

_, p_sw_ea = shapiro(exp_after)
_, p_sw_ca = shapiro(ctrl_after)
stat_lev_a, p_lev_a = levene(exp_after, ctrl_after)

md_ea = np.median(exp_after)
md_ca = np.median(ctrl_after)
iqr_ea = (np.percentile(exp_after, 25), np.percentile(exp_after, 75))
iqr_ca = (np.percentile(ctrl_after, 25), np.percentile(ctrl_after, 75))

print(f"  Экспериментальная (После): Md = {md_ea:.2f}, IQR = [{iqr_ea[0]:.2f}, {iqr_ea[1]:.2f}]")
print(f"  Контрольная (После):       Md = {md_ca:.2f}, IQR = [{iqr_ca[0]:.2f}, {iqr_ca[1]:.2f}]")
print(f"  Разница медиан:              ΔMd = {md_ea - md_ca:+.2f}")
print()
print(f"  Levene test: W = {stat_lev_a:.4f}, {interpret_pvalue(p_lev_a)}")
print()

if (p_sw_ea > 0.05) and (p_sw_ca > 0.05):
    equal_var_a = (p_lev_a > 0.05)
    stat_t, p_t = ttest_ind(exp_after, ctrl_after, equal_var=equal_var_a)
    d = cohens_d_independent(exp_after, ctrl_after)
    test_name = "Независимый t-тест" + (" (Классический)" if equal_var_a else " (Уэлча)")

    print("  ✅ Распределения нормальны.")
    print(f"  → {test_name}.")
    print(f"  Результат: t = {stat_t:.3f}, {interpret_pvalue(p_t)}")
    print(f"  Cohen's d = {d:.3f} - {interpret_cohens_d(d)} эффект")
    print()

    between_rows.append({
        'Сравнение': 'Результаты После',
        'Критерий': test_name,
        'Статистика': f"t = {stat_t:.3f}",
        'p-value': round(p_t, 4),
        'Размер эффекта': f"Cohen's d = {d:.3f} ({interpret_cohens_d(d)})",
        'Вывод': 'Различия значимы' if p_t < 0.05 else 'Различия НЕ значимы'
    })
else:
    stat_u, p_u = mannwhitneyu(exp_after, ctrl_after, alternative='two-sided')
    d_cliff = cliffs_delta(exp_after, ctrl_after)

    print("  ❌ Распределения НЕ нормальны.")
    print("  → Mann-Whitney U.")
    print(f"  Результат: U = {stat_u:.3f}, {interpret_pvalue(p_u)}")
    print(f"  Cliff's delta = {d_cliff:.3f} - {interpret_cliffs_delta(d_cliff)} эффект")
    print()

    between_rows.append({
        'Сравнение': 'Результаты После',
        'Критерий': 'Mann-Whitney U',
        'Статистика': f"U = {stat_u:.3f}",
        'p-value': round(p_u, 4),
        'Размер эффекта': f"Cliff's delta = {d_cliff:.3f} ({interpret_cliffs_delta(d_cliff)})",
        'Вывод': 'Различия значимы' if p_u < 0.05 else 'Разичия НЕ значимы'
    })

between_table = pd.DataFrame(between_rows)
show_table(between_table,
           title="СВОДНАЯ ТАБЛИЦА МЕЖГРУППОВЫХ СРАВНЕНИЙ",
           pvalue_cols=['p-value'],
           cmap='RdYlGn_r',
           precision=4)

# ═════════════════════════════════════════════════════════════════════════════
# 9. МАТЕМАТИЧЕСКАЯ МОДЕЛЬ (ANOVA / ANCOVA)
# ═════════════════════════════════════════════════════════════════════════════
print_section("5. МАТЕМАТИЧЕСКАЯ МОДЕЛЬ")

print_explanation(
    "ANOVA = анализ дисперсии. ANCOVA = ANOVA с ковариатой (контроль До)."
)

if STATSMODELS_AVAILABLE:
    df_long = pd.melt(
        df,
        id_vars=['Спортсмен', 'Группа'],
        value_vars=['До', 'После'],
        var_name='Время', value_name='Результат'
    )

    print_subsection("5a. Двухфакторная ANOVA (Группа x Время)")
    print_explanation(
        "Взаимодействие Группа x Время - это 'эффект методики'."
    )
    print("  Формула: Результат ~ Группа + Время + Группа:Время")
    print()

    model_ols = smf.ols('Результат ~ C(Группа) * C(Время)', data=df_long).fit()
    anova_table = sm.stats.anova_lm(model_ols, typ=2)
    print(anova_table)
    anova_df = anova_table.reset_index()
    anova_df.columns = ['Источник', 'Сумма квадратов', 'df', 'F', 'p-value']
    show_table(anova_df,
               title="ТАБЛИЦА ДВУХФАКТОРНОЙ ANOVA",
               gradient_cols=['F'],
               pvalue_cols=['p-value'],
               cmap='coolwarm',
               precision=4)

    print("  Проверка остатков модели (Breusch-Pagan test):")
    bp_test = het_breuschpagan(model_ols.resid, model_ols.model.exog)
    labels_bp = ['LM стат.', 'LM p-value', 'F стат.', 'F p-value']
    for lab, val in zip(labels_bp, bp_test):
        print(f"    {lab}: {val:.4f}")
    if bp_test[1] > 0.05:
        print("    ✅ Остатки гомоскедастичны.")
    else:
        print("    ⚠ Возможна гетероскедастичность.")
    print()

    print_subsection("5b. ANCOVA (контроль исходного уровня)")
    print_explanation(
        "Оценивает эффект группы ПРИ УЧЕТЕ исходных различий (До)."
    )
    print("  Формула: После ~ Группа + До")
    print()

    model_ancova = smf.ols('После ~ Группа + До', data=df).fit()
    ancova_df = pd.read_html(model_ancova.summary().tables[1].as_html(), header=0, index_col=0)[0]
    print(model_ancova.summary())
    ancova_df_reset = ancova_df.reset_index()
    ancova_df_reset.columns = ['Параметр', 'coef', 'std err', 't', 'P>|t|', '[0.025', '0.975]']
    show_table(ancova_df_reset,
               title="ТАБЛИЦА КОЭФФИЦИЕНТОВ ANCOVA",
               gradient_cols=['coef', 't'],
               pvalue_cols=['P>|t|'],
               cmap='coolwarm',
               precision=4)
else:
    print("   statsmodels не установлен. Установите: !pip install statsmodels")
    print()
    print("  Альтернатива - однофакторная ANOVA по изменениям:")
    stat_f, p_f = f_oneway(ctrl_change, exp_change)
    print(f"    F = {stat_f:.3f}, {interpret_pvalue(p_f)}")

# ═════════════════════════════════════════════════════════════════════════════
# 10. ВИЗУАЛИЗАЦИЯ (отдельные графики)
# ═════════════════════════════════════════════════════════════════════════════
print_section("6. ВИЗУАЛИЗАЦИЯ")

print_explanation(
    "Каждый график сохраняется отдельным файлом PNG."
)

# График 1: Скрипичные диаграммы
print_subsection("График 1: Скрипичные диаграммы До vs После")
fig, ax = plt.subplots(figsize=(10, 6))
df_melt = df.melt(id_vars=['Группа'], value_vars=['До', 'После'],
                  var_name='Время', value_name='Результат')
sns.violinplot(data=df_melt, x='Группа', y='Результат', hue='Время',
               split=True, palette={'До': '#9b59b6', 'После': '#2ecc71'}, ax=ax)
ax.set_title('Распределения результатов До и После по группам', fontweight='bold')
ax.legend(title='Время', loc='upper left')
plt.tight_layout()
save_figure("01_violin_before_after")

# График 2: Боксплоты изменений
print_subsection("График 2: Изменения (После - До) - боксплоты")
fig, ax = plt.subplots(figsize=(8, 6))
sns.boxplot(data=df, x='Группа', y='Изменение', palette=PALETTE, ax=ax)
ax.axhline(0, color='red', linestyle='--', alpha=0.7, label='Нет изменений (y=0)')
for i, g in enumerate([GROUP_CTRL, GROUP_EXP]):
    subset = df[df['Группа'] == g]['Изменение']
    med = subset.median()
    ax.text(i, med, f"Md={med:.1f}", ha='center', va='bottom',
            fontweight='bold', color='darkblue', fontsize=11)
ax.set_title('Изменения показателей по группам', fontweight='bold')
ax.legend()
plt.tight_layout()
save_figure("02_boxplot_changes")

# График 3: Индивидуальные траектории
print_subsection("График 3: Индивидуальные траектории (спагетти-график)")
fig, ax = plt.subplots(figsize=(10, 7))
for idx, row in ctrl_df.iterrows():
    ax.plot([0, 1], [row['До'], row['После']],
            color=PALETTE['Контрольная'], alpha=0.25, linewidth=0.8)
for idx, row in exp_df.iterrows():
    ax.plot([0, 1], [row['До'], row['После']],
            color=PALETTE['Экспериментальная'], alpha=0.25, linewidth=0.8)

ax.plot([0, 1],
        [ctrl_df['До'].mean(), ctrl_df['После'].mean()],
        color=PALETTE['Контрольная'], linewidth=4, marker='o', markersize=10,
        label=f'Контрольная (M)')
ax.plot([0, 1],
        [exp_df['До'].mean(), exp_df['После'].mean()],
        color=PALETTE['Экспериментальная'], linewidth=4, marker='s', markersize=10,
        label=f'Экспериментальная (M)')
ax.set_xticks([0, 1])
ax.set_xticklabels(['До', 'После'])
ax.set_ylabel('Результат')
ax.set_title('Индивидуальные траектории спортсменов', fontweight='bold')
ax.legend()
plt.tight_layout()
save_figure("03_spaghetti_plot")

# График 4: Столбчатые диаграммы медиан с IQR
print_subsection("График 4: Медианы До/После с IQR")
fig, ax = plt.subplots(figsize=(10, 6))
groups_plot = [GROUP_CTRL, GROUP_EXP]
x_pos = np.arange(len(groups_plot))
medians_before = [ctrl_df['До'].median(), exp_df['До'].median()]
medians_after = [ctrl_df['После'].median(), exp_df['После'].median()]

iqr_b_low = [ctrl_df['До'].quantile(0.25), exp_df['До'].quantile(0.25)]
iqr_b_high = [ctrl_df['До'].quantile(0.75), exp_df['До'].quantile(0.75)]
iqr_a_low = [ctrl_df['После'].quantile(0.25), exp_df['После'].quantile(0.25)]
iqr_a_high = [ctrl_df['После'].quantile(0.75), exp_df['После'].quantile(0.75)]

err_b = [[m - l for m, l in zip(medians_before, iqr_b_low)],
         [h - m for m, h in zip(medians_before, iqr_b_high)]]
err_a = [[m - l for m, l in zip(medians_after, iqr_a_low)],
         [h - m for m, h in zip(medians_after, iqr_a_high)]]

width = 0.35
ax.bar(x_pos - width/2, medians_before, width, yerr=err_b, capsize=5,
       label='До', color='#9b59b6', alpha=0.85, edgecolor='black')
ax.bar(x_pos + width/2, medians_after, width, yerr=err_a, capsize=5,
       label='После', color='#2ecc71', alpha=0.85, edgecolor='black')

for i, (b, a) in enumerate(zip(medians_before, medians_after)):
    ax.text(i - width/2, b + err_b[1][i] + 1, f"{b:.1f}",
            ha='center', va='bottom', fontweight='bold', fontsize=10)
    ax.text(i + width/2, a + err_a[1][i] + 1, f"{a:.1f}",
            ha='center', va='bottom', fontweight='bold', fontsize=10)

ax.set_xticks(x_pos)
ax.set_xticklabels(groups_plot)
ax.set_ylabel('Медиана с IQR')
ax.set_title('Медианы результатов До и После (с межквартильным размахом)', fontweight='bold')
ax.legend()
plt.tight_layout()
save_figure("04_barplot_medians_iqr")

# График 5: QQ-plot изменений (Контрольная)
print_subsection("График 5: QQ-plot изменений - Контрольная группа")
fig, ax = plt.subplots(figsize=(8, 6))
stats.probplot(ctrl_df['Изменение'], dist="norm", plot=ax)
ax.set_title('QQ-plot: нормальность распределения изменений (Контрольная)', fontweight='bold')
ax.get_lines()[0].set_markerfacecolor(PALETTE['Контрольная'])
ax.get_lines()[0].set_alpha(0.7)
plt.tight_layout()
save_figure("05_qqplot_control")

# График 6: QQ-plot изменений (Экспериментальная)
print_subsection("График 6: QQ-plot изменений - Экспериментальная группа")
fig, ax = plt.subplots(figsize=(8, 6))
stats.probplot(exp_df['Изменение'], dist="norm", plot=ax)
ax.set_title('QQ-plot: нормальность распределения изменений (Экспериментальная)', fontweight='bold')
ax.get_lines()[0].set_markerfacecolor(PALETTE['Экспериментальная'])
ax.get_lines()[0].set_alpha(0.7)
plt.tight_layout()
save_figure("06_qqplot_experimental")

# График 7: Гистограмма изменений (Контрольная)
print_subsection("График 7: Гистограмма изменений - Контрольная группа")
fig, ax = plt.subplots(figsize=(9, 6))
sns.histplot(ctrl_df['Изменение'], kde=True, color=PALETTE['Контрольная'],
             bins=12, edgecolor='black', alpha=0.7, ax=ax)
ax.axvline(ctrl_df['Изменение'].median(), color='red', linestyle='--', linewidth=2,
           label=f"Медиана = {ctrl_df['Изменение'].median():.1f}")
ax.axvline(0, color='black', linestyle='-', alpha=0.3)
ax.set_xlabel('Изменение (После - До)')
ax.set_ylabel('Частота')
ax.set_title('Распределение изменений (Контрольная)', fontweight='bold')
ax.legend()
plt.tight_layout()
save_figure("07_histogram_control")

# График 8: Гистограмма изменений (Экспериментальная)
print_subsection("График 8: Гистограмма изменений - Экспериментальная группа")
fig, ax = plt.subplots(figsize=(9, 6))
sns.histplot(exp_df['Изменение'], kde=True, color=PALETTE['Экспериментальная'],
             bins=12, edgecolor='black', alpha=0.7, ax=ax)
ax.axvline(exp_df['Изменение'].median(), color='red', linestyle='--', linewidth=2,
           label=f"Медиана = {exp_df['Изменение'].median():.1f}")
ax.axvline(0, color='black', linestyle='-', alpha=0.3)
ax.set_xlabel('зменение (После - До)')
ax.set_ylabel('Частота')
ax.set_title('Распределение изменений (Экспериментальная)', fontweight='bold')
ax.legend()
plt.tight_layout()
save_figure("08_histogram_experimental")

# График 9: Scatter До vs После
print_subsection("График 9: Зависимость После от До (точечный график)")
fig, ax = plt.subplots(figsize=(9, 7))
all_vals = df[['До', 'После']].values
min_val = all_vals.min() - 5
max_val = all_vals.max() + 5
ax.plot([min_val, max_val], [min_val, max_val], 'k--', alpha=0.5,
        label='Линия без изменений (y=x)')

ax.scatter(ctrl_df['До'], ctrl_df['После'],
           color=PALETTE['Контрольная'], s=80, alpha=0.7,
           edgecolors='black', linewidth=1, label='Контрольная')
ax.scatter(exp_df['До'], exp_df['После'],
           color=PALETTE['Экспериментальная'], s=80, alpha=0.7,
           marker='s', edgecolors='black', linewidth=1, label='Экспериментальная')

for g_name, g_df, col in [("Контрольная", ctrl_df, PALETTE['Контрольная']),
                            ("Экспериментальная", exp_df, PALETTE['Экспериментальная'])]:
    z = np.polyfit(g_df['До'], g_df['После'], 1)
    p_fn = np.poly1d(z)
    x_line = np.linspace(g_df['До'].min(), g_df['До'].max(), 100)
    ax.plot(x_line, p_fn(x_line), color=col, linewidth=2, linestyle='-.',
            label=f'{g_name}: тренд')
    r, p_r = pearsonr(g_df['До'], g_df['После'])
    print(f"     Корреляция До-После ({g_name}): r = {r:.3f}, {interpret_pvalue(p_r)}")

ax.set_xlabel('Результат До')
ax.set_ylabel('Результат После')
ax.set_title('Зависимость результатов После от До', fontweight='bold')
ax.legend(loc='upper left')
plt.tight_layout()
save_figure("09_scatter_before_after")

# График 10: Сводная панель эффектов
print_subsection("График 10: Сводная панель - медианы и доверительные интервалы")
fig, ax = plt.subplots(figsize=(10, 6))

categories = ['Контрольная\nДо', 'Контрольная\nПосле',
              'Экспериментальная\nДо', 'Экспериментальная\nПосле']
medians = [
    ctrl_df['До'].median(), ctrl_df['После'].median(),
    exp_df['До'].median(), exp_df['После'].median()
]
colors_bar = [PALETTE['Контрольная'], PALETTE['Контрольная'],
              PALETTE['Экспериментальная'], PALETTE['Экспериментальная']]
hatches = ['', '///', '', '///']

bars = ax.bar(categories, medians, color=colors_bar, edgecolor='black', alpha=0.85)
for bar, h in zip(bars, hatches):
    bar.set_hatch(h)

ci_data = [
    bootstrap_median_ci(ctrl_df['До'].values),
    bootstrap_median_ci(ctrl_df['После'].values),
    bootstrap_median_ci(exp_df['До'].values),
    bootstrap_median_ci(exp_df['После'].values)
]

for i, (med, (lo, hi)) in enumerate(zip(medians, ci_data)):
    ax.errorbar(i, med, yerr=[[med - lo], [hi - med]],
                fmt='none', ecolor='black', capsize=6, capthick=2)
    ax.text(i, hi + 1, f"{med:.1f}",
            ha='center', va='bottom', fontweight='bold', fontsize=10)

ax.set_ylabel('Медиана результата')
ax.set_title('Медианы с 95% доверительными интервалами (бутстрап)', fontweight='bold')
ax.set_ylim(bottom=min(medians) * 0.95)
plt.tight_layout()
save_figure("10_summary_effects")

# ═════════════════════════════════════════════════════════════════════════════
# 11. ИТОГОВЫЙ ВЫВОД
# ═════════════════════════════════════════════════════════════════════════════
print_section("7. ИТОГОВЫЙ ВЫВОД")

print("""
╔═══════════════════════════════════════════════════════════════════════════╗
║                        РЕЗЮМЕ ИССЛЕДОВАНИЯ                                 ║
╠════════════════════════════════════════════════════════════════════════════╣
║  1. Внутригрупповые изменения:                                           ║
║     • Проверено, изменились ли показатели в каждой группе отдельно.      ║
║     • Использованы параметрические (t-тест) или непараметрические        ║
║       (Wilcoxon) критерии в зависимости от нормальности.                ║
║     • Рассчитаны размеры эффектов (Cohen's d / Rosenthal's r).          ║
║                                                                          ║
║  2. Межгрупповые различия:                                               ║
║     • Сравнены изменения между группами (эффект методики).               ║
║     • Сравнены конечные результаты (После) между группами.             ║
║     • Использованы t-тест или Mann-Whitney U с Cliff's delta.          ║
║                                                                          ║
║  3. Математическая модель:                                              ║
║     • Построена ANOVA/ANCOVA для оценки взаимодействия Группа x Время. ║
║                                                                          ║
║  4. Визуализация:                                                       ║
║     • 10 отдельных графиков сохранены в виде PNG-файлов.               ║
║                                                                          ║
║  p < 0.05 означает, что наблюдаемые различия статистически             ║
║    достоверны, но НЕ доказывают причинно-следственную связь автоматически.║
║    Всегда интерпретируйте резултаты в контексте исследования!         ║
╚════════════════════════════════════════════════════════════════════════════╝
""")

print("✅ АНАЛИЗ УСПЕШНО ЗАВЕРШЕН!")
print("   Все графики сохранены в текущей директории.")
```

```python
# ============================================================
# 1. ГЕНЕРАЦИЯ СИНТЕТИЧЕСКИХ ДАННЫХ (спортивно-педагогическое исследование)
# ============================================================
np.random.seed(42)
n = 80  # общее число испытуемых

# Группы: контрольная и экспериментальная
group = np.array(['Экспериментальная'] * 40 + ['Контрольная'] * 40)

# Пол
gender = np.random.choice(['М', 'Ж'], size=n, p=[0.55, 0.45])

# Квалификация (качественная переменная)
qualification = np.random.choice(
    ['Начинающий', 'КМС', 'МС'], size=n, p=[0.3, 0.4, 0.3]
)

# азовые показатели (до эксперимента)
base_phys = np.random.normal(65, 10, n)   # физическая подготовленность
base_tech = np.random.normal(70, 8, n)    # техника
base_mot = np.random.normal(60, 12, n)    # мотивация

# После эксперимента: экспериментальная группа улучшается сильнее
post_phys = base_phys + np.where(
    group == 'Экспериментальная',
    np.random.normal(8, 3, n),
    np.random.normal(2, 3, n)
)
post_tech = base_tech + np.where(
    group == 'Экспериментальная',
    np.random.normal(7, 2.5, n),
    np.random.normal(1.5, 2.5, n)
)
post_mot = base_mot + np.where(
    group == 'Экспериментальная',
    np.random.normal(10, 4, n),
    np.random.normal(1, 4, n)
)

# Дополнительные показатели
coord = base_phys * 0.4 + base_tech * 0.3 + np.random.normal(0, 3, n)
speed = 100 - base_phys * 0.5 + np.random.normal(0, 4, n)

# Создание DataFrame
df = pd.DataFrame({
    'ID': range(1, n + 1),
    'Группа': group,
    'Пол': gender,
    'Квалификация': qualification,
    'Физ_подготовка_до': np.round(base_phys, 1),
    'Физ_подготовка_после': np.round(post_phys, 1),
    'Техника_до': np.round(base_tech, 1),
    'Техника_после': np.round(post_tech, 1),
    'Мотивация_до': np.round(base_mot, 1),
    'Мотивация_после': np.round(post_mot, 1),
    'Координация': np.round(coord, 1),
    'Быстрота': np.round(speed, 1),
})
df
df.to_csv('report.csv', index=False)
```

```python

"""
================================================================================
КОМПЛЕКСНЫЙ  АНАЛИЗ

================================================================================


ИНСТРУКЦИЯ ПО НАСТРОЙКЕ:
1. Загрузите ваш CSV-файл в среду Colab (через панель слева или код)
2. В разделе "КОНФИГУРАЦИЯ" укажите:
   - путь к файлу
   - название столбца с группами
   - пары столбцов "до" и "после"
   - количественные показатели
   - качественные переменные
3. Запустите все ячейки
================================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
import warnings
import os
warnings.filterwarnings('ignore')

plt.rcParams['figure.figsize'] = (14, 6)
plt.rcParams['font.size'] = 10
sns.set_theme(style="whitegrid", palette="muted")


# ============================================================================
# БЛОК ВСПОМОГАТЕЛЬНЫХ ФУНКЦИЙ
# ============================================================================

def robust_read_csv(filepath, **kwargs):
    """
    Надежная загрузка CSV с автоопределением кодировки, разделителя и наличия заголовка.
    Пробует несколько стратегий и выбирает ту, что сработала.
    """
    import io

    # Проверка существования файла
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Файл не найден: {filepath}. "
                                f"Пожалуйста, загрузите файл в среду выполнения Colab.")

    # Сначала читаем сырые байты для анализа
    with open(filepath, 'rb') as f:
        raw_bytes = f.read()

    # Список кодировок для попытки (в порядке приоритета)
    encodings_to_try = []

    # Пробуем автоопределение через charset_normalizer (если доступен)
    try:
        from charset_normalizer import detect
        detection = detect(raw_bytes)
        detected_enc = detection.get('encoding')
        if detected_enc:
            encodings_to_try.append(detected_enc)
            print(f"[Загрузка] Автоопределена кодировка: {detected_enc} "
                  f"(уверенность: {detection.get('confidence', 'N/A')})")
    except Exception:
        pass

    # Добавляем стандартные кодировки
    standard_encodings = ['utf-8-sig', 'utf-8', 'cp1251', 'windows-1251',
                          'latin-1', 'iso-8859-1', 'cp1252', 'koi8-r']
    for enc in standard_encodings:
        if enc not in encodings_to_try:
            encodings_to_try.append(enc)

    # Возможные разделители
    delimiters = [',', ';', '\t', '|']

    # Пробуем все комбинации кодировка + разделитель
    errors = []
    for encoding in encodings_to_try:
        for sep in delimiters:
            try:
                # Пробуем прочитать с этой кодировкой и разделителем
                text = raw_bytes.decode(encoding)
                df_test = pd.read_csv(io.StringIO(text), sep=sep, **kwargs)

                # Проверяем, что получился непустой DataFrame
                if df_test.empty:
                    continue

                # Проверяем, что количество колонок разумное
                if len(df_test.columns) < 2:
                    continue

                # Проверяем, что нет NaN в названиях колонок (часто признак проблемы)
                if df_test.columns.isna().any():
                    continue

                print(f"[Загрузка] УСПЕХ: кодировка={encoding}, разделитель='{sep}', "
                      f"строк={len(df_test)}, столбцов={len(df_test.columns)}")
                print(f"[Загрузка] Столбцы: {list(df_test.columns)}")
                return df_test

            except Exception as e:
                errors.append(f"  {encoding} + '{sep}': {type(e).__name__}: {str(e)[:80]}")
                continue

    # Если ничего не сработало — выводим диагнстику и падаем
    print("[ОШИБКА] Не удалось прочитать CSV-файл ни с одной из комбинаций кодировка+разделитель.")
    print("\nПопробованные комбинации:")
    for err in errors[:15]:
        print(err)
    if len(errors) > 15:
        print(f"  ... и еще {len(errors) - 15} ошибок")
    raise ValueError("Не удалось загрузить CSV-файл. Проверьте формат файла.")


def interpret_p(p):
    """Развернутая интерпретация p-значения."""
    if p < 0.001:
        return (f"p < 0.001 — различия высоко достоверны. "
                f"Вероятность случайного получения таких различий менее 0.1%. "
                f"Нулевая гипотеза об отсутствии различий отвергается.")
    elif p < 0.01:
        return (f"p = {p:.4f} — различия достоверны н уровне значимости 1%. "
                f"Нулевая гипотеза отвергается.")
    elif p < 0.05:
        return (f"p = {p:.4f} — различия статистически значимы на уровне 5%. "
                f"Нулевая гипотеза отвергается.")
    else:
        return (f"p = {p:.4f} — различия статистически НЕзначимы. "
                f"Нет оснований отвергать нулевую гипотезу об отсутствии различий. "
                f"Полученные различия могут быть обусловлены случайными факторами.")


def interpret_cohens_d(d):
    """Интерпретация размера эффекта Коэна."""
    abs_d = abs(d)
    direction = "в пользу первой группы" if d > 0 else "в пользу второй группы" if d < 0 else ""
    if abs_d < 0.2:
        return f"Cohen's d = {d:.3f} — пренебрежимо малый размер эффекта {direction}"
    elif abs_d < 0.5:
        return f"Cohen's d = {d:.3f} — малый размер эффекта {direction}"
    elif abs_d < 0.8:
        return f"Cohen's d = {d:.3f} — средний размер эффекта {direction}"
    elif abs_d < 1.2:
        return f"Cohen's d = {d:.3f} — большой размер эффекта {direction}"
    else:
        return f"Cohen's d = {d:.3f} — очень большой размер эффекта {direction}"


def interpret_correlation(r, p, method="Пирсона"):
    """Интерпретация коэффициента корреляции."""
    abs_r = abs(r)
    if abs_r < 0.1:
        strength = "практически отсутствует"
    elif abs_r < 0.3:
        strength = "слабая"
    elif abs_r < 0.5:
        strength = "умеренная"
    elif abs_r < 0.7:
        strength = "заметная (средняя)"
    elif abs_r < 0.9:
        strength = "высокая"
    else:
        strength = "очень высокая"
    direction = "прямая" if r > 0 else "обратная"
    sig = "достоверна" if p < 0.05 else "НЕдостоверна"
    return (f"Коэффициент корреляции {method}: r = {r:.3f}. "
            f"Связь {direction}, {strength}. "
            f"Статистическая значимость: {sig} (p = {p:.4f}).")


def check_normality(data, alpha=0.05):
    """Проверка нормальности с автовыбором теста."""
    n = len(data)
    if n <= 50:
        stat, p = stats.shapiro(data)
        test_name = "Шапиро-Уилка"
    else:
        z = (data - np.mean(data)) / np.std(data, ddof=1)
        stat, p = stats.kstest(z, 'norm')
        test_name = "Колмогорова-Смирнова"
    is_normal = p > alpha
    return is_normal, p, test_name, stat


def auto_two_sample_test(group1, group2, paired=False, alpha=0.05):
    """Автоматический выбор и выполнение двухвыборочного теста."""
    g1, g2 = np.array(group1), np.array(group2)
    if paired:
        diff = g1 - g2
        is_normal, p_norm, test_name, stat_norm = check_normality(diff, alpha)
        if is_normal:
            stat, p = stats.ttest_rel(g1, g2)
            test_used = "парный t-критерий Стьюдента"
            interpretation = (f"Распределение разниц нормально ({test_name}: p = {p_norm:.4f}). "
                            f"Применен параметрический парный t-критерий.")
        else:
            stat, p = stats.wilcoxon(g1, g2)
            test_used = "T-критерий Вилкоксона"
            interpretation = (f"Распределение разниц отклоняется от нормального "
                            f"({test_name}: p = {p_norm:.4f}). Применен непараметрический "
                            f"T-критерий Вилкоксона.")
        return {
            'test': test_used,
            'statistic': stat,
            'p': p,
            'interpretation': interpretation,
            'normality_p': p_norm,
            'normality_test': test_name
        }
    else:
        n1, n2 = len(g1), len(g2)
        norm1, p1, t1, s1 = check_normality(g1, alpha)
        norm2, p2, t2, s2 = check_normality(g2, alpha)
        lev_stat, lev_p = stats.levene(g1, g2)
        equal_var = lev_p > alpha
        if norm1 and norm2:
            stat, p = stats.ttest_ind(g1, g2, equal_var=equal_var)
            test_used = "t-критерий Стьюдента" + (" (равные дисперсии)" if equal_var else " (неравные дисперсии, поправка Уэлча)")
            interpretation = (f"Обе выборки распределены нормально "
                            f"(группа 1: {t1} p = {p1:.4f}; группа 2: {t2} p = {p2:.4f}). "
                            f"Гомогенность диперсий: {'подтверждена' if equal_var else 'НЕ подтверждена'} "
                            f"(Levene p = {lev_p:.4f}). Применен {test_used}.")
            pooled_std = np.sqrt(((n1-1)*np.var(g1, ddof=1) + (n2-1)*np.var(g2, ddof=1)) / (n1+n2-2))
            cohens_d = (np.mean(g1) - np.mean(g2)) / pooled_std if pooled_std > 0 else 0
        else:
            stat, p = stats.mannwhitneyu(g1, g2, alternative='two-sided')
            test_used = "U-критерий Манна-Уитни"
            interpretation = (f"Хотя бы одна выборка НЕ распределена нормально "
                            f"(группа 1: {t1} p = {p1:.4f}; группа 2: {t2} p = {p2:.4f}). "
                            f"Применен непараметрический U-критерий Манна-Уитни.")
            cohens_d = None
        return {
            'test': test_used,
            'statistic': stat,
            'p': p,
            'interpretation': interpretation,
            'cohens_d': cohens_d,
            'normality': (norm1, norm2),
            'levene_p': lev_p,
            'equal_var': equal_var
        }


def interpret_anova(f_stat, p, eta_sq, group_names):
    """Интерпретация результатов ANOVA."""
    if p < 0.05:
        eta_interpret = (
            "пренебрежимо малый" if eta_sq < 0.01 else
            "малый" if eta_sq < 0.06 else
            "средний" if eta_sq < 0.14 else "большой"
        )
        return (f"ANOVA выявила статистически значимые различия между группами "
                f"({', '.join(group_names)}): F = {f_stat:.3f}, {interpret_p(p)} "
                f"Размер эффекта eta-squared = {eta_sq:.4f} ({eta_interpret}). "
                f"Это означает, что {eta_sq*100:.1f}% дисперсии зависимой переменной "
                f"объясняется принадлежностью к группам.")
    else:
        return (f"ANOVA НЕ выявила статистически значимых различий между группами "
                f"({', '.join(group_names)}): F = {f_stat:.3f}, {interpret_p(p)}")


# ============================================================================
# КОНФИГУРАЦИЯ — измените под ваши данные
# ============================================================================

from google.colab import files
uploaded = files.upload()

# Определяем имя загруженного файла
file_name = list(uploaded.keys())[0]
print(f"\n Загружен файл: {file_name}")

# Загружаем данные в зависимости от расширения
try:
    if file_name.endswith('.csv'):
        df = pd.read_csv(file_name)
    else:
        df = pd.read_excel(file_name)
    print("✅ Данные успешно загружены!")
except Exception as e:
    print(f"❌ Ошибка загрузки: {e}")
    print("Попробуйте сохранить файл в формате .xlsx или .csv и загрузить заново.")
    raise


# Путь к CSV-файлу. Если None — будут сгенерированы демо-данные.
# Примеры пуей для Colab:
#   CSV_PATH = '/content/my_data.csv'
#   CSV_PATH = '/content/data.csv'
CSV_PATH = '/content/report.csv'

# Столбец с группами (например, 'Группа', 'Group')
GROUP_COL = 'Группа'

# Список пар столбцов "до" и "после" для парного анализа
# Формат: [('название_до', 'название_после', 'человекочитаемое_название'), ...]
PAIRS_BEFORE_AFTER = [
    ('Физ_подготовка_до', 'Физ_подготовка_после', 'Физическая подготовленность'),
    ('Техника_до',        'Техника_после',        'Техника'),
    ('Мотивация_до',      'Мотивация_после',      'Мотивация'),
]

# Дополнительные количественные показатели (не входящие в пары до/после)
ADDITIONAL_NUMERIC = ['Координация', 'Быстрота']

# Качественные переменные для анализа таблиц сопряженности
CATEGORICAL_COLS = ['Пол', 'Квалификация']

# Зависимая переменная для регрессии
REGRESSION_TARGET = 'Дин_физ'

# Предикторы для множественной регрессии
REGRESSION_PREDICTORS = ['Физ_подготовка_до', 'Техника_до', 'Мотивация_до']

# Уровень значимости
ALPHA = 0.05


# ============================================================================
# 1. ЗАГРУЗКА ИЛИ ГЕНЕРАЦИЯ ДАННЫХ
# ============================================================================
print("=" * 75)
print(" КОМПЛЕКСНЫЙ СТАТИСТИЧЕСКИЙ АНАЛИЗ")
print("")
print("=" * 75)





if CSV_PATH:
    print(f"\n[1] ЗАГРУЗКА ДАННЫХ ИЗ: {CSV_PATH}")
    #df = robust_read_csv(CSV_PATH)
    print(f"\n    Данные успешно загружены: {df.shape[0]} строк x {df.shape[1]} столбцов")

    # Автоматически определяем числовые столбцы
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    if 'ID' in numeric_cols:
        numeric_cols.remove('ID')

    # Проверяем наличие указанных столбцов
    all_specified = []
    for before, after, _ in PAIRS_BEFORE_AFTER:
        all_specified.extend([before, after])
    all_specified.extend(ADDITIONAL_NUMERIC)
    all_specified.extend(CATEGORICAL_COLS)
    all_specified.append(GROUP_COL)

    missing = [c for c in all_specified if c not in df.columns]
    if missing:
        print(f"\n[ПРЕДУПРЕЖДЕНИЕ] Следующие столбцы из конфигурации не найдены в CSV: {missing}")
        print(f"    Доступные столбцы: {list(df.columns)}")
else:
    print("\n[1] CSV НЕ УКАЗАН — ГЕНЕРИРУЮТСЯ ДЕМОНСТРАЦИОННЫЕ ДАННЫЕ")
    np.random.seed(42)
    n = 80
    group = np.array(['Экспериментальная'] * 40 + ['Контрольная'] * 40)
    gender = np.random.choice(['М', 'Ж'], size=n, p=[0.55, 0.45])
    qualification = np.random.choice(['Начинающий', 'КМС', 'МС'], size=n, p=[0.3, 0.4, 0.3])
    base_phys = np.random.normal(65, 10, n)
    base_tech = np.random.normal(70, 8, n)
    base_mot = np.random.normal(60, 12, n)
    post_phys = base_phys + np.where(group == 'Экспериментальная', np.random.normal(8, 3, n), np.random.normal(2, 3, n))
    post_tech = base_tech + np.where(group == 'Экспериментальная', np.random.normal(7, 2.5, n), np.random.normal(1.5, 2.5, n))
    post_mot = base_mot + np.where(group == 'Экспериментальная', np.random.normal(10, 4, n), np.random.normal(1, 4, n))
    coord = base_phys * 0.4 + base_tech * 0.3 + np.random.normal(0, 3, n)
    speed = 100 - base_phys * 0.5 + np.random.normal(0, 4, n)
    df = pd.DataFrame({
        'ID': range(1, n+1),
        GROUP_COL: group,
        'Пол': gender,
        'Квалификация': qualification,
        'Физ_подготовка_до': np.round(base_phys, 1),
        'Физ_подготовка_после': np.round(post_phys, 1),
        'Техника_до': np.round(base_tech, 1),
        'Техника_после': np.round(post_tech, 1),
        'Мотивация_до': np.round(base_mot, 1),
        'Мотивация_после': np.round(post_mot, 1),
        'Координация': np.round(coord, 1),
        'Быстрота': np.round(speed, 1),
    })
    for before, after, _ in PAIRS_BEFORE_AFTER:
        df[f"Дин_{before.split('_')[0].lower()}"] = df[after] - df[before]

print(f"\nПервые 5 строк:")
print(df.head())
print(f"\nТипы данных:")
print(df.dtypes)


# ============================================================================
# 2. ОПИСАТЕЛЬНАЯ СТАТИСТИКА
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП I. ОПИСАТЕЛЬНАЯ СТАТИСТИКА")
print("=" * 75)

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if 'ID' in numeric_cols:
    numeric_cols.remove('ID')

desc = df[numeric_cols].agg(['count', 'mean', 'median', 'std', 'min', 'max'])
desc.loc['std_error'] = df[numeric_cols].std() / np.sqrt(df[numeric_cols].count())
desc.loc['var'] = df[numeric_cols].var()
print("\n[2.1] Числовые характеристики выборки (M ± SD, медиана, мин-макс):")
print(np.round(desc.T, 2).to_string())

print("\n[2.2] ИНТЕРПРЕТАЦИЯ:")
for col in numeric_cols:
    mean_v = df[col].mean()
    std_v = df[col].std()
    cv = std_v / mean_v * 100 if mean_v != 0 else 0
    variability = (
        "низкая вариабельность" if cv < 10 else
        "средняя вариабельность" if cv < 20 else "высокая вариабельность"
    )
    print(f"    • {col}: M = {mean_v:.2f}, SD = {std_v:.2f}, CV = {cv:.1f}% — {variability}")

fig, axes = plt.subplots(2, min(len(PAIRS_BEFORE_AFTER), 3), figsize=(5*min(len(PAIRS_BEFORE_AFTER), 3), 10))
if len(PAIRS_BEFORE_AFTER) == 1:
    axes = np.array([[axes[0]], [axes[1]]])
for idx, (before, after, name) in enumerate(PAIRS_BEFORE_AFTER[:3]):
    axes[0, idx].hist(df[before], bins=15, alpha=0.6, label='До', color='steelblue', edgecolor='black')
    axes[0, idx].hist(df[after], bins=15, alpha=0.6, label='После', color='coral', edgecolor='black')
    axes[0, idx].set_title(f'{name}\n(распределение)')
    axes[0, idx].set_xlabel('Баллы')
    axes[0, idx].set_ylabel('Частота')
    axes[0, idx].legend()
    dyn_col = f"Дин_{before.split('_')[0].lower()}"
    if dyn_col in df.columns:
        sns.boxplot(data=df, x=GROUP_COL, y=dyn_col, ax=axes[1, idx], palette='Set2')
        axes[1, idx].set_title(f'Динамика: {name}')
        axes[1, idx].set_ylabel('Прирост')
plt.tight_layout()

plt.show()



# ============================================================================
# 3. ПРОВЕРКА НОРМАЛЬНОСТИ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП II. ПРОВЕРКА НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ")
print("=" * 75)

normality_results = []
print("\n[3.1] Результаты тестов на нормальность:")
for col in numeric_cols:
    is_normal, p, test_name, stat = check_normality(df[col].dropna(), ALPHA)
    normality_results.append({
        'Показатель': col,
        'Тест': test_name,
        'Статистика': round(stat, 4),
        'p-value': round(p, 4),
        'Нормальность': 'Да' if is_normal else 'Нет'
    })
    status = "РАСПРЕДЕЛЕНИЕ НОРМАЛЬНОЕ" if is_normal else "РАСПРЕДЕЛЕНИЕ ОТЛИЧАЕТСЯ ОТ НОРМАЛЬНОГО"
    print(f"    • {col}: {test_name} = {stat:.4f}, p = {p:.4f} — {status}")

norm_df = pd.DataFrame(normality_results)
print("\n[3.2] СВОДНАЯ ТАБЛИЦА:")
print(norm_df.to_string(index=False))

qq_cols = [before for before, _, _ in PAIRS_BEFORE_AFTER[:3]]
fig, axes = plt.subplots(1, len(qq_cols), figsize=(6*len(qq_cols), 5))
if len(qq_cols) == 1:
    axes = [axes]
for idx, col in enumerate(qq_cols):
    stats.probplot(df[col], dist="norm", plot=axes[idx])
    axes[idx].set_title(f'Q-Q plot: {col}')
    axes[idx].get_lines()[0].set_markerfacecolor('steelblue')
    axes[idx].get_lines()[1].set_color('coral')
plt.tight_layout()

plt.show()



# ============================================================================
# 4. ГОМОГЕННОСТЬ ДИСПЕРСИЙ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП III. ПРОВЕРКА ГОМОГЕННОСТИ ДИСПЕРСИЙ")
print("=" * 75)

if GROUP_COL in df.columns and df[GROUP_COL].nunique() == 2:
    groups = df[GROUP_COL].unique()
    g1_name, g2_name = groups[0], groups[1]
    g1_data = df[df[GROUP_COL] == g1_name]
    g2_data = df[df[GROUP_COL] == g2_name]
    print(f"\n[4.1] Критерий Левене ля 2 групп ({g1_name} vs {g2_name}):")
    for col in [before for before, _, _ in PAIRS_BEFORE_AFTER]:
        stat, p = stats.levene(g1_data[col], g2_data[col])
        result = "ДИСПЕРСИИ ОДНОРОДНЫ" if p > ALPHA else "ДИСПЕРСИИ ГЕТЕРОГЕННЫ"
        print(f"    • {col}: Levene = {stat:.4f}, p = {p:.4f} — {result}")
        if p > ALPHA:
            print(f"      → Можно применять классический t-критерий Стьюдента с равными дисперсиями.")
        else:
            print(f"      → Следует использовать t-критерий с поправкой Уэлча (неравные дисперсии).")

if len(CATEGORICAL_COLS) > 0:
    for cat_col in CATEGORICAL_COLS:
        if df[cat_col].nunique() >= 3:
            print(f"\n[4.2] Критерий Бартлетта по '{cat_col}' (≥3 группы):")
            grps = [grp[PAIRS_BEFORE_AFTER[0][0]].values for _, grp in df.groupby(cat_col)]
            if all(len(g) > 1 for g in grps):
                stat, p = stats.bartlett(*grps)
                result = "ДИСПЕРСИИ ОДНОРОДНЫ" if p > ALPHA else "ДИСПЕРСИИ ГЕТЕРОГЕННЫ"
                print(f"    • {PAIRS_BEFORE_AFTER[0][2]}: Bartlett = {stat:.4f}, p = {p:.4f} — {result}")


# ============================================================================
# 5. СРАВНИТЕЛЬНЫЙ АНАЛИЗ ДВУХ ГРУПП
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП IV. СРАВНИТЕЛЬНЫЙ АНАЛИЗ ДВУХ ГРУПП")
print("=" * 75)

if GROUP_COL in df.columns and df[GROUP_COL].nunique() == 2:
    groups = df[GROUP_COL].unique()
    g1_name, g2_name = groups[0], groups[1]
    g1_data = df[df[GROUP_COL] == g1_name]
    g2_data = df[df[GROUP_COL] == g2_name]
    comparison_vars = [before for before, _, _ in PAIRS_BEFORE_AFTER] + \
                      [after for _, after, _ in PAIRS_BEFORE_AFTER] + \
                      [f"Дин_{before.split('_')[0].lower()}" for before, _, _ in PAIRS_BEFORE_AFTER] + \
                      ADDITIONAL_NUMERIC
    comparison_vars = [c for c in comparison_vars if c in df.columns]
    print(f"\n[5.1] СРАВНЕНИЕ ГРУПП '{g1_name}' (n={len(g1_data)}) и '{g2_name}' (n={len(g2_data)}):\n")
    comparison_results = []
    for col in comparison_vars:
        print(f"  ► ПОКАЗАТЕЛЬ: {col}")
        print(f"    Средние: {g1_name} = {g1_data[col].mean():.2f} ± {g1_data[col].std():.2f}; "
              f"{g2_name} = {g2_data[col].mean():.2f} ± {g2_data[col].std():.2f}")
        res = auto_two_sample_test(g1_data[col], g2_data[col], paired=False, alpha=ALPHA)
        print(f"    {res['interpretation']}")
        print(f"    Результат: {res['test']}: статистика = {res['statistic']:.3f}, {interpret_p(res['p'])}")
        if res['cohens_d'] is not None:
            print(f"    {interpret_cohens_d(res['cohens_d'])}")
        print()
        comparison_results.append({
            'Показатель': col,
            'Тест': res['test'],
            'Статистика': round(res['statistic'], 3),
            'p-value': round(res['p'], 4),
            'Cohen_d': round(res['cohens_d'], 3) if res['cohens_d'] is not None else None
        })
    dyn_cols = [c for c in comparison_vars if 'Дин_' in c]
    if len(dyn_cols) > 0:
        fig, axes = plt.subplots(1, len(dyn_cols), figsize=(6*len(dyn_cols), 5))
        if len(dyn_cols) == 1:
            axes = [axes]
        for idx, col in enumerate(dyn_cols):
            sns.violinplot(data=df, x=GROUP_COL, y=col, ax=axes[idx], palette='Set2', inner='box')
            axes[idx].set_title(col)
        plt.tight_layout()

        plt.show()

else:
    print(f"\n[5.1] ПРОПУЩЕНО: групповая переменная '{GROUP_COL}' не найдена или имеет ≠2 групп.")


# ============================================================================
# 6. ДИСПЕРСИОННЫЙ АНАЛИЗ (ANOVA)
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП V. ДИСПЕРСИОННЫЙ АНАЛИЗ (ANOVA)")
print("=" * 75)

anova_targets = []
if GROUP_COL in df.columns and df[GROUP_COL].nunique() > 2:
    anova_targets.append((GROUP_COL, df[GROUP_COL].unique()))
for cat_col in CATEGORICAL_COLS:
    if cat_col in df.columns and df[cat_col].nunique() >= 3:
        anova_targets.append((cat_col, df[cat_col].unique()))

for factor_col, factor_levels in anova_targets:
    print(f"\n[6.1] Однофакторный ANOVA по фактору '{factor_col}':")
    for before, after, name in PAIRS_BEFORE_AFTER:
        dyn_col = f"Дин_{before.split('_')[0].lower()}"
        target_col = dyn_col if dyn_col in df.columns else after
        grps = [grp[target_col].values for _, grp in df.groupby(factor_col)]
        if all(len(g) > 1 for g in grps):
            f_stat, p_anova = stats.f_oneway(*grps)
            ss_between = sum(len(g) * (np.mean(g) - df[target_col].mean())**2 for g in grps)
            ss_total = ((df[target_col] - df[target_col].mean())**2).sum()
            eta_sq = ss_between / ss_total if ss_total > 0 else 0
            print(f"\n  ► {name} (динамика):")
            print(f"    {interpret_anova(f_stat, p_anova, eta_sq, factor_levels)}")
            levels = df[factor_col].unique()
            n_comp = len(levels) * (len(levels) - 1) // 2
            alpha_corr = ALPHA / n_comp if n_comp > 0 else ALPHA
            print(f"    Попарные сравнения (поправка Бонферрони, alpha = {alpha_corr:.4f}):")
            for i in range(len(levels)):
                for j in range(i+1, len(levels)):
                    g1 = df[df[factor_col] == levels[i]][target_col]
                    g2 = df[df[factor_col] == levels[j]][target_col]
                    t, p = stats.ttest_ind(g1, g2)
                    sig = "*" if p < alpha_corr else "нс"
                    print(f"      {levels[i]} vs {levels[j]}: t = {t:.3f}, p = {p:.4f} ({sig})")

if len(anova_targets) > 0:
    factor_col = anova_targets[0][0]
    dyn_col = f"Дин_{PAIRS_BEFORE_AFTER[0][0].split('_')[0].lower()}"
    if dyn_col in df.columns:
        fig, ax = plt.subplots(figsize=(10, 5))
        hue = GROUP_COL if GROUP_COL in df.columns else None
        sns.boxplot(data=df, x=factor_col, y=dyn_col, hue=hue, palette='Set2', ax=ax)
        ax.set_title(f'Динамика по фактору "{factor_col}"')
        plt.tight_layout()

        plt.show()



# ============================================================================
# 7. ПАРНЫЙ АНАЛИЗ (до vs после)
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП VI. АНАЛИЗ ПОВТОРНЫХ ИЗМЕРЕНИЙ (до vs после)")
print("=" * 75)

print("\n[7.1] Сравнение показателей ДО и ПОСЛЕ эксперимента:\n")
for before, after, name in PAIRS_BEFORE_AFTER:
    print(f"  ► {name} ({before} vs {after}):")
    print(f"    Общая выборка: До = {df[before].mean():.2f} ± {df[before].std():.2f}; "
          f"После = {df[after].mean():.2f} ± {df[after].std():.2f}")
    res_all = auto_two_sample_test(df[before], df[after], paired=True, alpha=ALPHA)
    print(f"    {res_all['interpretation']}")
    print(f"    Результат: {res_all['test']}: статистика = {res_all['statistic']:.3f}, {interpret_p(res_all['p'])}")
    if GROUP_COL in df.columns:
        for grp_name in df[GROUP_COL].unique():
            grp_df = df[df[GROUP_COL] == grp_name]
            res_grp = auto_two_sample_test(grp_df[before], grp_df[after], paired=True, alpha=ALPHA)
            print(f"    • Группа '{grp_name}': {res_grp['test']}, статистика = {res_grp['statistic']:.3f}, "
                  f"p = {res_grp['p']:.4f}")
    print()

fig, axes = plt.subplots(1, len(PAIRS_BEFORE_AFTER), figsize=(6*len(PAIRS_BEFORE_AFTER), 5))
if len(PAIRS_BEFORE_AFTER) == 1:
    axes = [axes]
for idx, (before, after, name) in enumerate(PAIRS_BEFORE_AFTER):
    plot_df = pd.melt(df, id_vars=[GROUP_COL] if GROUP_COL in df.columns else ['ID'],
                      value_vars=[before, after], var_name='Этап', value_name=name)
    plot_df['Этап'] = plot_df['Этап'].map({before: 'До', after: 'После'})
    hue = GROUP_COL if GROUP_COL in df.columns else None
    sns.pointplot(data=plot_df, x='Этап', y=name, hue=hue, ax=axes[idx], palette='Set2', capsize=0.1)
    axes[idx].set_title(f'{name}\n(средние ± SEM)')
plt.tight_layout()

plt.show()



# ============================================================================
# 8. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП VII. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ")
print("=" * 75)

corr_candidates = [before for before, _, _ in PAIRS_BEFORE_AFTER] + ADDITIONAL_NUMERIC
corr_vars = [c for c in corr_candidates if c in df.columns]

if len(corr_vars) >= 2:
    print("\n[8.1] Матрица корреляций Пирсона:")
    pearson_corr = df[corr_vars].corr(method='pearson')
    print(np.round(pearson_corr, 3).to_string())
    print("\n[8.2] Матрица корреляций Спирмена:")
    spearman_corr = df[corr_vars].corr(method='spearman')
    print(np.round(spearman_corr, 3).to_string())
    print("\n[8.3] ИНТЕРПРЕТАЦИЯ ЗНАЧИМЫХ КОРРЕЛЯЦИЙ (Пирсон):")
    for i in range(len(corr_vars)):
        for j in range(i+1, len(corr_vars)):
            r, p = stats.pearsonr(df[corr_vars[i]], df[corr_vars[j]])
            if p < ALPHA:
                print(f"    • {corr_vars[i]} x {corr_vars[j]}:")
                print(f"      {interpret_correlation(r, p, 'Пирсона')}")
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    mask = np.triu(np.ones_like(pearson_corr, dtype=bool))
    sns.heatmap(pearson_corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
                center=0, vmin=-1, vmax=1, ax=axes[0], square=True)
    axes[0].set_title('Корреляции Пирсона')
    sns.heatmap(spearman_corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
                center=0, vmin=-1, vmax=1, ax=axes[1], square=True)
    axes[1].set_title('Корреляции Спирмена')
    plt.tight_layout()

    plt.show()

    sig_pairs = []
    for i in range(len(corr_vars)):
        for j in range(i+1, len(corr_vars)):
            r, p = stats.pearsonr(df[corr_vars[i]], df[corr_vars[j]])
            if abs(r) > 0.3 and p < ALPHA:
                sig_pairs.append((corr_vars[i], corr_vars[j], r, p))
    if len(sig_pairs) > 0:
        n_sig = min(len(sig_pairs), 4)
        fig, axes = plt.subplots(1, n_sig, figsize=(5*n_sig, 5))
        if n_sig == 1:
            axes = [axes]
        for idx, (x, y, r, p) in enumerate(sig_pairs[:n_sig]):
            sns.regplot(data=df, x=x, y=y, ax=axes[idx],
                        scatter_kws={'color': 'steelblue'}, line_kws={'color': 'coral'})
            axes[idx].set_title(f'{x} vs {y}\nr = {r:.3f}, p = {p:.4f}')
        plt.tight_layout()

        plt.show()



# ============================================================================
# 9. РЕГРЕССИОННЫЙ АНАЛИЗ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП VIII. РЕГРЕССИОННЫЙ АНАЛИЗ")
print("=" * 75)

if len(corr_vars) >= 2:
    x_col, y_col = corr_vars[0], corr_vars[1]
    X_s = df[[x_col]].values
    y_s = df[y_col].values
    model_s = LinearRegression().fit(X_s, y_s)
    r2_s = model_s.score(X_s, y_s)
    n_s = len(y_s)
    f_reg_s = (r2_s / 1) / ((1 - r2_s) / (n_s - 2)) if r2_s < 1 else np.inf
    p_reg_s = 1 - stats.f.cdf(f_reg_s, 1, n_s - 2) if f_reg_s != np.inf else 0.0
    print(f"\n[9.1] Простая линейная регрессия: {y_col} = f({x_col})")
    print(f"    Уравнение: Y = {model_s.intercept_:.3f} + {model_s.coef_[0]:.3f} * X")
    print(f"    R-squared = {r2_s:.4f} — модель объясняет {r2_s*100:.1f}% дисперсии зависимой переменной.")
    print(f"    F = {f_reg_s:.3f}, p = {p_reg_s:.4f}")
    if p_reg_s < ALPHA:
        print(f"    → Регрессионная модель статистически значима. "
              f"Прдиктор '{x_col}' достоверно влияет на '{y_col}'.")
    else:
        print(f"    → Регрессионная модель статистически НЕзначима.")

preds = [p for p in REGRESSION_PREDICTORS if p in df.columns]
if REGRESSION_TARGET in df.columns and len(preds) > 0:
    X_m = df[preds].values
    y_m = df[REGRESSION_TARGET].values
    model_m = LinearRegression().fit(X_m, y_m)
    r2_m = model_m.score(X_m, y_m)
    n_m, k_m = len(y_m), len(preds)
    f_reg_m = (r2_m / k_m) / ((1 - r2_m) / (n_m - k_m - 1)) if r2_m < 1 else np.inf
    p_reg_m = 1 - stats.f.cdf(f_reg_m, k_m, n_m - k_m - 1) if f_reg_m != np.inf else 0.0
    print(f"\n[9.2] Множественная регрессия: {REGRESSION_TARGET} = f({', '.join(preds)})")
    print(f"    Коэффициенты:")
    for name, coef in zip(preds, model_m.coef_):
        direction = "положительное" if coef > 0 else "отрицательное"
        print(f"      • {name}: {coef:.4f} ({direction} влияние)")
    print(f"    Intercept = {model_m.intercept_:.4f}")
    print(f"    R-squared = {r2_m:.4f} — совокупность предикторов объясняет {r2_m*100:.1f}% дисперсии.")
    print(f"    F = {f_reg_m:.3f}, p = {p_reg_m:.4f}")
    if p_reg_m < ALPHA:
        print(f"    → Множественная регрессионная модель в целом статистически значима.")
    else:
        print(f"    → Множественная регрессионная модель в целом статистически НЕзначима.")
    print(f"    Интерпретация: при изменении предикторов на 1 балл результат изменяется "
          f"согласно коэффициентам регрессии.")

fig, ax = plt.subplots(figsize=(10, 6))
y_pred_s = model_s.predict(X_s)
ax.scatter(df[x_col], df[y_col], alpha=0.6, color='steelblue', edgecolors='black')
ax.plot(df[x_col], y_pred_s, color='coral', linewidth=2)
ax.set_xlabel(x_col)
ax.set_ylabel(y_col)
ax.set_title(f'Простая линейная регрессия\nY = {model_s.intercept_:.2f} + {model_s.coef_[0]:.2f}X, R² = {r2_s:.3f}')
plt.tight_layout()

plt.show()



# ============================================================================
# 10. КАЧЕСТВЕННЫЕ ПЕРЕМЕННЫЕ (хи-квадрат)
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП IX. АНАЛИЗ КАЧЕСТВЕННЫХ ПЕРЕМЕННЫХ")
print("=" * 75)

if GROUP_COL in df.columns and len(CATEGORICAL_COLS) > 0:
    for cat_col in CATEGORICAL_COLS:
        if cat_col in df.columns:
            contingency = pd.crosstab(df[GROUP_COL], df[cat_col])
            print(f"\n[10.1] Таблица сопряженности: '{GROUP_COL}' x '{cat_col}':")
            print(contingency)
            chi2, p_chi, dof, expected = stats.chi2_contingency(contingency)
            n_total = contingency.sum().sum()
            cramers_v = np.sqrt(chi2 / (n_total * (min(contingency.shape) - 1))) if min(contingency.shape) > 1 else 0
            print(f"\n[10.2] Критерий хи-квадрат:")
            print(f"    chi2 = {chi2:.4f}, df = {dof}, p = {p_chi:.4f}")
            print(f"    {interpret_p(p_chi)}")
            v_interp = (
                "отсутствует" if cramers_v < 0.1 else
                "слабая" if cramers_v < 0.3 else
                "умеренная" if cramers_v < 0.5 else "заметная"
            )
            print(f"    V Крамера = {cramers_v:.4f} — связь {v_interp}.")
            if p_chi < ALPHA:
                print(f"    → Между группой и '{cat_col}' существует статистически значимая связь.")
            else:
                print(f"    → Связь между группой и '{cat_col}' статистически НЕзначима.")
    if len(CATEGORICAL_COLS) > 0:
        fig, ax = plt.subplots(figsize=(8, 4))
        cat_col = CATEGORICAL_COLS[0]
        contingency = pd.crosstab(df[GROUP_COL], df[cat_col])
        sns.heatmap(contingency, annot=True, fmt='d', cmap='YlOrRd', ax=ax)
        ax.set_title(f'Таблица сопряженности: {GROUP_COL} x {cat_col}')
        plt.tight_layout()

        plt.show()



# ============================================================================
# 11. ФАКТОРНЫЙ АНАЛИЗ (PCA)
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП X. ФАКТОРНЫЙ АНАЛИЗ (метод главных компонент)")
print("=" * 75)

fa_vars = [c for c in corr_vars if c in df.columns]
if len(fa_vars) >= 3:
    X_fa = df[fa_vars].dropna().values
    scaler = StandardScaler()
    X_fa_scaled = scaler.fit_transform(X_fa)
    n_comp = min(len(fa_vars), 5)
    pca = PCA(n_components=n_comp)
    pca.fit(X_fa_scaled)
    print("\n[11.1] Объясненная дисперсия компонентами:")
    for i, (var, ratio) in enumerate(zip(pca.explained_variance_, pca.explained_variance_ratio_)):
        print(f"    PC{i+1}: eigenvalue = {var:.3f}, доля дисперсии = {ratio:.3f} ({ratio*100:.1f}%)")
    cumulative = np.cumsum(pca.explained_variance_ratio_)
    n_80 = np.argmax(cumulative >= 0.80) + 1
    print(f"\n    Для объяснения 80% дисперсии достаточно {n_80} компонент().")
    print(f"    Общая объясненная дисперсия первых 3 компонент: {cumulative[min(2, n_comp-1)]*100:.1f}%")
    loadings = pd.DataFrame(
        pca.components_[:3].T,
        columns=['PC1', 'PC2', 'PC3'],
        index=fa_vars
    )
    print("\n[11.2] Факторные нагрузки (первые 3 компоненты):")
    print(np.round(loadings, 3).to_string())
    print("\n    Интерпретация факторных нагрузок:")
    for var in fa_vars:
        pc_loads = loadings.loc[var]
        max_pc = pc_loads.abs().idxmax()
        max_val = pc_loads[max_pc]
        print(f"    • {var}: наибольшая нагрузка на {max_pc} = {max_val:.3f} "
              f"({'сильная' if abs(max_val) > 0.7 else 'умеренная' if abs(max_val) > 0.4 else 'слабая'} связь)")
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    axes[0].bar(range(1, n_comp+1), pca.explained_variance_ratio_, color='steelblue', edgecolor='black')
    axes[0].set_xlabel('Главная компонента')
    axes[0].set_ylabel('Доля объясненной дисперсии')
    axes[0].set_title('Scree plot')
    axes[0].set_xticks(range(1, n_comp+1))
    scores = pca.transform(X_fa_scaled)
    group_vals = df[GROUP_COL].values if GROUP_COL in df.columns else np.array([''] * len(scores))
    colors = {'Экспериментальная': 'coral', 'Контрольная': 'steelblue'}
    for grp in np.unique(group_vals):
        mask = group_vals == grp
        c = colors.get(grp, 'gray')
        axes[1].scatter(scores[mask, 0], scores[mask, 1], c=c, label=grp, alpha=0.7, edgecolors='black')
    for i, var in enumerate(fa_vars):
        axes[1].arrow(0, 0, pca.components_[0, i]*2.5, pca.components_[1, i]*2.5,
                      color='red', head_width=0.05)
        axes[1].text(pca.components_[0, i]*2.7, pca.components_[1, i]*2.7, var, color='red', fontsize=8)
    axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
    axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
    axes[1].set_title('Биплот PCA')
    axes[1].legend()
    plt.tight_layout()

    plt.show()



# ============================================================================
# 12. КЛАСТЕРНЫЙ АНАЛИЗ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП XI. КЛАСТЕРНЫЙ АНАЛИЗ")
print("=" * 75)

if len(fa_vars) >= 3:
    linked = linkage(X_fa_scaled, method='ward')
    kmeans = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_fa_scaled)
    df['Кластер'] = kmeans.labels_
    print("\n[12.1] Иерархическая кластеризация (метод Уорда):")
    print("    Построена дендрограмма для визуальной оценки числа кластеров.")
    print("\n[12.2] K-means кластеризация (k=3):")
    cluster_counts = pd.Series(kmeans.labels_).value_counts().sort_index()
    for clust, count in cluster_counts.items():
        print(f"    • Кластер {clust}: {count} испытуемых")
    print("\n[12.3] Характеристика кластеров (средние значения показателей):")
    cluster_means = df.groupby('Кластер')[fa_vars].mean()
    print(np.round(cluster_means, 2).to_string())
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    dendrogram(linked, ax=axes[0], color_threshold=0.7*max(linked[:, 2]), no_labels=True)
    axes[0].set_title('Дендрограмма (метод Уорда)')
    axes[0].set_xlabel('Испытуемые')
    axes[0].set_ylabel('Евклидово расстояние')
    scatter = axes[1].scatter(scores[:, 0], scores[:, 1], c=kmeans.labels_, cmap='Set1', edgecolors='black', alpha=0.8)
    axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
    axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
    axes[1].set_title('Кластеры K-means (k=3) в пространстве PC1-PC2')
    plt.colorbar(scatter, ax=axes[1], label='Кластер')
    plt.tight_layout()

    plt.show()



# ============================================================================
# 13. ИТОГОВАЯ СВОДКА
# ============================================================================
print("\n" + "=" * 75)
print("ИТОГОВАЯ СВОДКА ПРОВЕДЕННОГО АНАЛИЗА")
print("=" * 75)

print("""
Проведенные этапы анализа:

1. ОПИСАТЕЛЬНАЯ СТАТИСТИКА
   - Рассчитаны средние, медианы, стандартные отклонения, ошибки среднего,
     коэффициенты вариации для всех количественных показателей.
   - Дана интерпретация вариабельности данных.

2. ПРОВЕРКА НОРМАЛЬНОСТИ
   - Автоматический выбор теста: Шапиро-Уилка (n ≤ 50) или
     Колмогорова-Смирнова (n > 50).
   - Q-Q графики для визуальной оценки соответствия нормальному
     распределению.

3. ГОМОГЕННОСТЬ ДИСПЕРСИЙ
   - Критерий Левене для сравнения двух групп.
   - Критерий Бартлетта для сравнения трех и более групп.
   - Автоматический выбор поправки Уэлча при нарушении гомогенности.

4. СРАВНИТЕЛЬНЫЙ АНАЛИЗ ГРУПП
   - Автоматический выбор между t-критерием Стьюдента и
     U-критерием Манна-Уитни на основе проверки нормальности.
   - Расчет размера эффекта Коэна (Cohen's d) для параметрических тестов.
   - Подробная текстовая интерпретация каждого результата.

5. ДИСПЕРСИОННЫЙ АНАЛИЗ (ANOVA)
   - Однофакторный ANOVA для факторов с ≥3 уровнями.
   - Расчет eta-squared (размер эффекта).
   - Попарные сравнения с поправкой Бонферрони.

6. АНАЛИЗ ПОВТОРНЫХ ИЗМЕРЕНИЙ
   - Автоматический выбор между парным t-критерием и
     T-критерием Вилкоксона.
   - Анализ общей выборки и отдельно по группам.

7. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ
   - Матрицы корреляций Пирсона и Спирмена.
   - Интерпретация силы и направления связи.
   - Scatter plots с линией регрессии для значимых пар.

8. РЕГРЕССИОННЫЙ АНАЛИЗ
   - Простая линейная регрессия.
   - Множественная регрессия с несколькими предикторами.
   - Оценка R² и F-теста с текстовой интерпретацией.

9. АНАЛИЗ КАЧЕСТВЕННЫХ ПЕРЕМЕННЫХ
   - Таблицы сопряжености.
   - Критерий хи-квадрат Пирсона.
   - V Крамера с интерпретацией силы связи.

10. ФАКТОРНЫЙ АНАЛИЗ
    - Метод главных компонент (PCA).
    - Scree plot, биплот, факторные нагрузки.
    - Определение числа компонент для объяснения 80% дисперсии.

11. КЛАСТЕРНЫЙ АНАЛИЗ
    - Иерархическая кластеризация (дендрограмма).
    - K-means кластеризация.
    - Визуализация кластеров в PCA-пространстве.
    - Описание характеристик каждого кластера.
""")
print("=" * 75)
print("АНАЛИЗ ЗАВЕРШЕН.")
print("=" * 75)
```
